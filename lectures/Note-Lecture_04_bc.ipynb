1. Tóm tắt nội dung chính của Lecture 04 (Portfolio Theory and Practice I)

Bài học hôm nay xoay quanh hai câu hỏi nền tảng của lý thuyết đầu tư tài chính:

Làm sao để đo lường lợi nhuận (Return) và rủi ro (Risk)?
Làm sao để tối ưu hóa sự kết hợp giữa các tài sản để đạt lợi ích lớn nhất?
Các nội dung chính được chia làm 3 phần cốt lõi:

Lợi nhuận đầu tư (Return on Investments):

- Holding Period Return (HPR): Lợi nhuận thực tế kiếm được trong một khoảng thời gian nắm giữ tài sản (bao gồm chênh lệch giá mua-bán và các nguồn thu nhập định kỳ như cổ tức/tiền lãi).

- Lợi nhuận kỳ vọng (Expected Return): Trung bình trọng số của các kịch bản có thể xảy ra trong tương lai (bùm nổ, bình thường, suy thoái) dựa trên xác suất.

- Lợi nhuận vượt mức (Excess Return): Phần lợi nhuận kiếm thêm được so với khi gửi tiền vào tài sản phi rủi ro (ví dụ: gửi tiết kiệm ngân hàng hoặc trái phiếu chính phủ).

Rủi ro đầu tư (Risk in Investments):

Chúng ta đo lường rủi ro bằng sự biến động (Volatility) - cụ thể là Phương sai (Variance) và Độ lệch chuẩn (Standard Deviation) của lợi nhuận.

Khi phân phối lợi nhuận không chuẩn (bị lệch hoặc có đuôi dày - Skewness & Kurtosis), độ lệch chuẩn không còn phản ánh đủ rủi ro nữa, lúc đó phải dùng thêm các chỉ số như VaR (Value at Risk) hay CVaR để đo lường rủi ro thua lỗ nặng ở vùng đuôi.

Lý thuyết danh mục đầu tư & Sự đánh đổi (Portfolio Theory & Trade-off):

Mức độ ngại rủi ro (Risk Aversion) & Hàm độ thỏa dụng (Utility Function): Mỗi nhà đầu tư có một khẩu vị rủi ro khác nhau (ngại rủi ro, trung dung, hay thích rủi ro).

- Hàm số $U = E(r) - \frac{1}{2}A\sigma^2$$U = E(r) - \frac{1}{2}A\sigma^2$ giúp tính toán xem một danh mục mang lại bao nhiêu "sự hài lòng" cho họ.

- Hiệu ứng đa dạng hóa (Diversification):

Khi kết hợp các tài sản có mối tương quan (Correlation) thấp hoặc ngược chiều nhau, chúng ta có thể giảm đáng kể rủi ro (độ lệch chuẩn) của cả danh mục mà không làm giảm đi mức lợi nhuận trung bình một cách tương ứng. Đây chính là nguyên lý "không bỏ hết trứng vào một giỏ".

- Đường biên hiệu quả (Efficient Frontier) & GMVP: Tìm ra danh mục có rủi ro thấp nhất có thể (Global Minimum Variance Portfolio) và tập hợp các danh mục tối ưu nhất.

2. Ví dụ thực tế gần gũi dành cho sinh viên BBA

Để dễ hình dung, bạn hãy tạm quên đi những con số tài chính phức tạp trên sàn chứng khoán. Hãy nghĩ về một việc rất quen thuộc với sinh viên BBA: Quản lý thời gian học tập và làm thêm trong một học kỳ.

Giả sử bạn có 2 hoạt động chính để phân bổ thời gian:

- Dự án khởi nghiệp nhóm (Tài sản rủi ro cao - Alpha): Nếu dự án thành công xuất sắc, bạn nhận điểm A+ và có cơ hội gọi vốn thực tế (Lợi nhuận cực cao).
Nhưng nếu nhóm lục đục hoặc ý tưởng thất bại, bạn có thể bị điểm F và mất trắng công sức (Rủi ro rất lớn).

- Làm trợ giảng cho Giáo sư (Tài sản rủi ro thấp/Phi rủi ro - Beta): Công việc này nhẹ nhàng, mỗi tháng nhận lương cố định và chắc chắn được điểm rèn luyện tốt (Lợi nhuận vừa phải nhưng cực kỳ an toàn).

Áp dụng lý thuyết bài học vào ví dụ này:

- Holding Period Return (Lợi nhuận nắm giữ):

Cuối kỳ, tổng kết lại bạn nhận được bao nhiêu kiến thức, điểm số và tiền lương từ hai hoạt động trên.

- Expected Return & Risk (Lợi nhuận kỳ vọng & Rủi ro):

Trước khi bắt đầu kỳ học, bạn tự tính toán: "Nếu mình đầu tư 80% thời gian vào Dự án khởi nghiệp, trung bình mình sẽ đạt kết quả thế nào? Độ biến động (stress, mất ngủ, rủi ro tạch môn) là bao nhiêu?"
Risk Aversion (Khẩu vị rủi ro):

- Nếu bạn là người ngại rủi ro (A > 0): Bạn sẽ chọn dành 70% thời gian làm trợ giảng an toàn, chỉ dành 30% cho dự án khởi nghiệp.
- Nếu bạn là người thích mạo hiểm (A < 0): Bạn sẵn sàng dốc 90% sức lực vào khởi nghiệp để tìm kiếm cơ hội bứt phá.

- Diversification (Đa dạng hóa): Nếu bạn chỉ làm đúng 1 dự án khởi nghiệp duy nhất, bạn đối mặt với rủi ro 100% thất bại. Nhưng nếu bạn tham gia vào 2 dự án ở 2 lĩnh vực khác nhau (ví dụ: một dự án Công nghệ và một dự án F&B), nếu ngành F&B gặp khó khăn thì dự án Công nghệ vẫn gánh được. Nhờ đó, tổng "rủi ro điểm kém" của bạn giảm đi đáng kể mà cơ hội học hỏi vẫn rất cao.

### Return on Investments
*   Total Return (Holding Period Return)
    *   The total return realized on an investment in a given period and has two components:
        *   Capital gains in the relevant period,
        *   Income generated during the period (Dividend, coupon, rent, etc.)

$$TR = \frac{P_1 - P_0 + I}{P_0}$$

| TR  | Total return                               |
| :-- | :----------------------------------------- |
| P1  | The price of the asset at the end of the period    |
| P0  | The price of the asset at the beginning of the period |
| I   | Income generated during the holding period |

*   Total return calculations do not take the return on reinvestment of the income earned during the period into account.
*   The holding period return is not a standardized value in terms of time.

### 1. Ý nghĩa của Slide

Slide giải thích rằng khi bạn bỏ tiền ra đầu tư một tài sản trong một khoảng thời gian, tổng lợi nhuận bạn thu về gồm **2 thành phần chính**:

1. **Tăng giá vốn (Capital Gains):** Giá trị tài sản tăng lên so với lúc bạn mua ($P_1 - P_0$).
2. **Thu nhập định kỳ (Income Generated - $I$):** Tiền bạn nhận được trong lúc nắm giữ tài sản (ví dụ: cổ tức từ cổ phiếu, tiền lãi từ trái phiếu, hoặc tiền cho thuê nhà).

#### Công thức:

$$TR = \frac{P_1 - P_0 + I}{P_0}$$

* $P_0$: Giá mua lúc đầu (Beginning price)
* $P_1$: Giá bán/giá hiện tại ở cuối kỳ (Ending price)
* $I$: Thu nhập nhận được trong kỳ (Income)

### 2. Ví dụ thực tế dễ hiểu cho sinh viên

**Kịch bản: Mua xe máy đi chạy Grab / Đi học**

* Đầu năm nhất, bạn mua lại một chiếc xe Wave cũ với giá **10.000.000 VNĐ** ($P_0$).
* Trong suốt 1 năm, bạn dùng xe chạy Grab những lúc rảnh và kiếm về tổng cộng **3.000.000 VNĐ** tiền lời sau khi trừ chi phí xăng xe ($I$).
* Đầu năm hai, bạn không dùng nữa và bán lại chiếc xe đó cho sinh viên khóa dưới với giá **8.500.000 VNĐ** ($P_1$).

**Tính tổng lợi nhuận (HPR) của bạn:**

1. **Lợi nhuận từ chênh lệch giá (Capital Gain):**

$$8.500.000 - 10.000.000 = -1.500.000 \text{ VNĐ (Lỗ 1,5 triệu do xe mất giá)}$$


2. **Thu nhập trong kỳ (Income):**

$$+3.000.000 \text{ VNĐ (Lời 3 triệu từ tiền chạy Grab)}$$


3. **Tổng tiền lời thực tế:**

$$-1.500.000 + 3.000.000 = +1.500.000 \text{ VNĐ}$$



**Tỷ lệ lợi nhuận ($TR$):**


$$TR = \frac{8.500.000 - 10.000.000 + 3.000.000}{10.000.000} = \frac{1.500.000}{10.000.000} = 15\%$$

> **Kết luận:** Dù giá trị chiếc xe bị giảm (chênh lệch giá âm), nhưng nhờ có dòng tiền kiếm được từ việc chạy Grab ($I$), tổng lợi nhuận cả năm của bạn vẫn đạt **15%**.

### 3. Hai lưu ý quan trọng ở cuối Slide

* **Không tính tái đầu tư (No reinvestment):** Công thức này giả định số tiền thu nhập ($I$) nhận được trong kỳ nằm yên một chỗ, không tính đến việc bạn đem $I$ đi gửi tiết kiệm hay đầu tư tiếp để sinh lời kép.


Hiểu đơn giản: công thức TR chỉ tính "tiền bạn nhận được", chứ không tính "tiền đó có được đem đi sinh lời tiếp hay không".

Ví dụ: Bạn mua cổ phiếu, giữa kỳ nhận được 4 đô tiền cổ tức. Công thức TR chỉ cộng 4 đô này vào thẳng, coi như bạn bỏ nó vào ngăn kéo cất đó.

Nhưng thực tế, nếu bạn đem ngay 4 đô đó đi mua thêm cổ phiếu hoặc gửi tiết kiệm, nó sẽ tiếp tục sinh ra lời (lãi kép). Công thức TR không "thấy" được phần lời thêm này, nên con số TR thực ra hơi thấp hơn một chút so với lợi nhuận thật nếu bạn biết tái đầu tư khéo léo.

* **Không chuẩn hóa theo thời gian (Not standardized in terms of time):** HPR chỉ tính cho đúng khoảng thời gian bạn nắm giữ. Nếu bạn nắm giữ 3 tháng được 10% thì HPR là 10%; nếu nắm giữ 2 năm mới được 10% thì HPR vẫn ghi là 10% (chưa quy đổi ra lãi suất theo năm - Annualized Return).

Hiểu đơn giản: TR không quan tâm bạn giữ tài sản bao lâu, nó chỉ trả lời "tổng cộng bạn lời bao nhiêu %" trong suốt thời gian đó — dù là 1 tháng hay 10 năm.

Ví dụ dễ hình dung:

	* Bạn A giữ cổ phiếu 3 tháng, lời 10% → TR = 10%
	* Bạn B giữ cổ phiếu 2 năm, cũng lời 10% → TR = 10%

Nhìn vào TR thì thấy hai người "lời bằng nhau", nhưng thực chất bạn A lời nhanh hơn rất nhiều so với bạn B (vì chỉ mất 3 tháng để đạt 10%, trong khi bạn B mất tới 2 năm). Vì vậy, muốn so sánh công bằng hai khoản đầu tư có thời gian giữ khác nhau, người ta phải quy đổi TR này về "lợi nhuận theo năm" (Annualized Return) bằng công thức riêng — giống như cách tính trong ví dụ Alpha Corp. ở slide sau đó.

Tóm lại: hai lưu ý này đang nhắc bạn rằng HPR chỉ là con số thô ban đầu, cần cẩn thận khi dùng nó để so sánh giữa các khoản đầu tư khác nhau, vì nó chưa tính đến "tiền đẻ ra tiền" và "thời gian giữ dài hay ngắn".

### Return on Investments

$$
TR_{\text{annual}}=(1+TR)^{1/N}-1
$$

An investor bought shares of Alpha Corp. for USD 52 per share on April 1, 2024. The investor received a dividend of USD 1.80 per share and sold the shares for USD 58 on January 1, 2026. Calculate the holding-period return and the annualized total return.

$$
TR=\frac{P_1-P_0+D}{P_0}
=\frac{58-52+1.80}{52}
=0.15=15.00\%
$$

The holding period is 21 months, so $N=21/12=1.75$ years.

$$
TR_{\text{annual}}=(1+0.15)^{1/1.75}-1
=0.0831=8.31\%
$$

### Return on Investments

On July 1, 2023, Investor Beta purchased a government bond with a face value of €1,000 for €960. The bond pays €25 every June 30 and December 31. Beta sold the bond for €985 immediately after receiving the December 31, 2024 coupon. Calculate the holding-period return and the annualized total return.

Beta received three coupon payments during the 18-month holding period:

$$
C=3(25)=75
$$

$$
TR=\frac{P_1-P_0+C}{P_0}
=\frac{985-960+75}{960}
=0.1042=10.42\%
$$

The holding period is 18 months, so $N=1.5$ years.

$$
TR_{\text{annual}}=(1+0.1041667)^{1/1.5}-1
=0.0683=6.83\%
$$

### Return on Investments: Expected Return

The expected return is the probability-weighted average return across possible states:

$$
E(r)=\sum_s p(s)r(s).
$$

A share of Gamma Corp. costs USD 50 today. Its possible prices and dividends one year from now are:

| Market state | Probability | Price $P_1$ | Dividend $D_1$ | Return $r(s)$ | $p(s)r(s)$ |
|---|---:|---:|---:|---:|---:|
| Boom | 10% | 68.00 | 2.00 | 40.00% | 4.00% |
| Growth | 20% | 61.00 | 1.50 | 25.00% | 5.00% |
| Normal | 40% | 54.00 | 1.00 | 10.00% | 4.00% |
| Recession | 20% | 47.00 | 0.50 | −5.00% | −1.00% |
| Depression | 10% | 40.00 | 0.00 | −20.00% | −2.00% |
| Total | 100% |  |  |  | 10.00% |

For example, the return in a boom is

$$
r(\text{Boom})=\frac{68-50+2}{50}=0.40=40.00\%.
$$

Therefore,

$$
\begin{aligned}
E(r)
&=0.10(0.40)+0.20(0.25)+0.40(0.10)\\
&\quad+0.20(-0.05)+0.10(-0.20)\\
&=0.10=10.00\%.
\end{aligned}
$$

### Return on Investments: Excess Return

Excess return is the return on a risky asset minus the risk-free rate:

$$
r_{\text{excess}}(s)=r(s)-r_f.
$$

Continue with the Gamma Corp. example from cell 8 and assume a one-year risk-free rate of 4%.

| Market state | Probability | Gamma return | Excess return | $p(s)[r(s)-r_f]$ |
|---|---:|---:|---:|---:|
| Boom | 10% | 40.00% | 36.00% | 3.60% |
| Growth | 20% | 25.00% | 21.00% | 4.20% |
| Normal | 40% | 10.00% | 6.00% | 2.40% |
| Recession | 20% | −5.00% | −9.00% | −1.80% |
| Depression | 10% | −20.00% | −24.00% | −2.40% |
| Total | 100% |  |  | 6.00% |

For example, in a boom,

$$
r_{\text{excess}}(\text{Boom})=0.40-0.04=0.36.
$$

The expected excess return is

$$
\begin{aligned}
E(r-r_f)
&=0.10(0.36)+0.20(0.21)+0.40(0.06)\\
&\quad+0.20(-0.09)+0.10(-0.24)\\
&=0.06=6.00\%.
\end{aligned}
$$

Equivalently,

$$
E(r-r_f)=E(r)-r_f=10.00\%-4.00\%=6.00\%.
$$

Chào An Ho Truong! Mình sẽ giải thích phần Excess Return (Tỷ suất sinh lời vượt trội) trong slide này một cách dễ hiểu nhé 😊

1. Khái niệm cơ bản
Excess Return là phần lợi nhuận "dôi ra" khi bạn đầu tư vào tài sản rủi ro (ví dụ: cổ phiếu) so với việc gửi tiền vào nơi an toàn, không rủi ro (ví dụ: trái phiếu chính phủ).

Công thức:
r_(excess)(s)= r(s)−r_f

Trong đó:

	* r(s): lợi nhuận của tài sản rủi ro ở trạng thái thị trường s
	* r_f: lãi suất phi rủi ro (risk-free rate) — ở đây là 4%/năm

Ý nghĩa đơn giản: Nếu bạn chấp nhận rủi ro để đầu tư, bạn mong muốn được "đền bù" nhiều hơn so với gửi tiết kiệm an toàn. Phần đền bù đó chính là excess return.

2. Ví dụ minh họa từ bảng Gamma Corp.
Giả sử công ty Gamma Corp. có 5 kịch bản kinh tế có thể xảy ra, mỗi kịch bản có xác suất và lợi nhuận riêng:

| Trạng thái | Xác suất | Lợi nhuận Gamma | Excess return |
|---|---|---|---|
| Boom (bùng nổ) | 10% | 40% | 36% |
| Growth (tăng trưởng) | 20% | 25% | 21% |
| Normal (bình thường) | 40% | 10% | 6% |
| Recession (suy thoái) | 20% | -5% | -9% |
| Depression (khủng hoảng) | 10% | -20% | -24% |

Cách tính Excess Return cho từng dòng: chỉ cần lấy lợi nhuận trừ đi 4% (risk-free rate).

Ví dụ với kịch bản Boom:
r_(excess)("Boom")= 0.40 − 0.04 = 0.36 = 36%

Tương tự cho Recession:
r_(excess)("Recession")=−0.05 − 0.04 =−0.09 =−9%

3. Tính kỳ vọng Excess Return — E(r − r_f)
Đây là phần quan trọng nhất: bạn muốn biết trung bình (kỳ vọng) của excess return là bao nhiêu, bằng cách nhân xác suất mỗi kịch bản với excess return tương ứng rồi cộng lại:

E(r − r_f)= 0.10(0.36)+ 0.20(0.21)+ 0.40(0.06)+ 0.20(−0.09)+ 0.10(−0.24)
Tính từng phần:

	* 0.10 × 0.36 = 0.036
	* 0.20 × 0.21 = 0.042
	* 0.40 × 0.06 = 0.024
	* 0.20 ×(−0.09)=−0.018
	* 0.10 ×(−0.24)=−0.024

Cộng lại: 0.036 + 0.042 + 0.024 − 0.018 − 0.024 = 0.06 = 6%

4. Cách tính nhanh hơn (tương đương)
Thay vì tính từng kịch bản, ta có thể dùng công thức rút gọn:

E(r − r_f)= E(r)−r_f
Vì E(r) (kỳ vọng lợi nhuận của Gamma, tính sẵn từ trước) = 10%, và r_f = 4%, nên:

E(r − r_f)= 10% − 4% = 6%

### Risk in Investments
*   One of the main functions of financial markets is to enable economic agents to transfer their risks to willing counterparties or to assume the risks of others.
*   Financial risks are generally associated with uncertainty about the outcome of an investment.
*   Accordingly, volatility (variance) in returns is the main risk criterion.
*   Risk measures are not limited to volatility:
    *   [VaR (Value at Risk)](https://pages.stern.nyu.edu/~adamodar/pdfiles/papers/VAR.pdf)
        *   Measures the potential loss in value of a risky asset or portfolio over a defined period at a given confidence level.
    *   [Expected Shortfall or CVaR (Conditional Value at Risk)](https://mayerkrebs.com/what-is-expected-shortfall-cvar/)
        *   The expected loss given that the loss is greater than or equal to the VaR.
        *   Quantifies the losses that might be encountered in the tail.
    *   [Semivariance](https://fastercapital.com/content/Downside-deviation--Navigating-Downside-Deviation-through-Semivariance.html)
        *   Measures the dispersion of all observations that fall below the mean or target value of a set of data.
    *   Sortino Ratio
        *   Measures average return above a target per unit of downside deviation
        *   Downside deviation is the square root of target semivariance

### 3. Các thước đo rủi ro ngoài Độ biến động (Risk Measures beyond Volatility)

#### ① VaR (Value at Risk - Giá trị chịu rủi ro)

* **Nội dung:** Đo lường mức **thua lỗ tối đa có thể gặp phải** trong một khoảng thời gian nhất định, với một mức độ tin cậy cụ thể (thường là $95\%$ hoặc $99\%$).
* **Ví dụ ngắn:** "Với độ tin cậy $95\%$, trong 1 tháng tới mức lỗ tối đa của bạn là 10 triệu VNĐ". Tức là có $95\%$ khả năng bạn chỉ lỗ tối đa 10 triệu, nhưng vẫn có $5\%$ khả năng bạn lỗ nhiều hơn 10 triệu.

#### ② Expected Shortfall / CVaR (Conditional Value at Risk)

* **Nội dung:** Khi kịch bản tồi tệ nhất rơi vào $5\%$ còn lại (vượt quá ngưỡng VaR), **mức lỗ trung bình lúc đó là bao nhiêu?** Chỉ số này giúp đo lường rủi ro thua lỗ cực nặng ở phần "đuôi" phân phối (tail risk).
* **Ví dụ ngắn:** Nếu bạn lỡ rơi vào $5\%$ kịch bản xui xẻo nhất ở trên, CVaR cho biết trung bình bạn sẽ mất **18 triệu VNĐ** chứ không dừng lại ở 10 triệu.

#### ③ Semivariance (Bán phương sai)

* **Nội dung:** Phương sai thông thường tính độ lệch cả khi giá tăng lẫn khi giá giảm. Nhưng nhà đầu tư chỉ sợ giá giảm! Semivariance **chỉ đo độ biến động của các mức lợi nhuận nằm dưới mức trung bình / mục tiêu**.
* **Ví dụ ngắn:** Giá cổ phiếu tăng mạnh $30\%$ là tin tốt, nhưng phương sai vẫn coi đó là "biến động/rủi ro". Semivariance sẽ bỏ qua đợt tăng $30\%$ này và chỉ tập trung tính toán rủi ro của những ngày cổ phiếu sụt giảm.

#### ④ Sortino Ratio (Tỷ số Sortino)

* **Nội dung:** Đo lường phần **lợi nhuận vượt mức trên mỗi đơn vị rủi ro chiều giảm (Downside deviation)**.
* **Ví dụ ngắn:** Giúp trả lời câu hỏi: *"Với mỗi 1% rủi ro bị rớt giá/thua lỗ, danh mục đầu tư này đền bù lại cho bạn bao nhiêu % lợi nhuận vượt kỳ vọng?"*

### 1. Ý nghĩa & Vai trò của Phân phối chuẩn trong Đầu tư

* **Điểm bắt đầu của lý thuyết:** Các mô hình định giá tài sản và lý thuyết danh mục đầu tư hiện đại thường bắt đầu bằng giả định lợi nhuận đầu tư tuân theo **phân phối chuẩn (hình quả chuông)**.

* **Đo lường rủi ro chính xác:** Nhờ giả định này, **Độ lệch chuẩn ($\sigma$)** trở thành thước đo rủi ro hoàn hảo và duy nhất. Bạn chỉ cần biết lợi nhuận trung bình và độ lệch chuẩn là có thể hình dung toàn bộ bức tranh rủi ro.

### 2. Các đặc tính cốt lõi của Phân phối chuẩn

Slide trình bày 4 đặc tính toán học trọng tâm:

#### ① Được xác định hoàn toàn bởi Giá trị trung bình ($\mu$) và Phương sai ($\sigma^2$)

* Nghĩa là bạn chỉ cần 2 tham số này là vẽ được trọn vẹn đường cong phân phối.

* **Phân phối chuẩn chuẩn hóa (Standard Normal Distribution):** Có trung bình $\mu = 0$ và phương sai $\sigma^2 = 1$.

#### ② Giá trị trung bình ($\mu$) là Tham số vị trí (Location Parameter)

* $\mu$ quyết định **đỉnh quả chuông nằm ở đâu** trên trục hoành.

* Thay đổi $\mu$ sẽ dịch chuyển toàn bộ quả chuông sang trái (nếu $\mu$ giảm) hoặc sang phải (nếu $\mu$ tăng) mà không làm thay đổi hình dáng quả chuông (như hình minh họa ở giữa).

#### ③ Độ lệch chuẩn ($\sigma$) là Tham số tỷ lệ/độ rộng (Scale Parameter)

* $\sigma$ quyết định **độ dẹt hay nhọn** của quả chuông (độ biến động).
* $\sigma$ càng nhỏ, quả chuông càng cao nhọn (dữ liệu tập trung gần trung bình, ít biến động).
* $\sigma$ càng lớn, quả chuông càng thấp bẹt (dữ liệu phân tán rộng, biến động mạnh/rủi ro cao) (như hình minh họa bên phải).

```python
# If needed, uncomment and run once:
# %pip install -U yfinance curl_cffi pandas numpy matplotlib scipy ipywidgets

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from scipy.stats import norm, skew, kurtosis
from IPython.display import display, HTML, clear_output
import datetime

# --- 1. Data Download and Preparation ---
tickers = ['^SP500TR', 'QQQ', 'EEM', 'GC=F', 'AMD', 'AMZN', 'F', 'JPM', 'NVDA', 'PFE', 'TGT', 'XOM']
ticker_map = {
    '^SP500TR': 'SP500', 'QQQ': 'NDQ100', 'EEM': 'EEM', 'GC=F': 'Gold Futures',
    'AMD': 'AMD', 'AMZN': 'AMZN', 'F': 'FORD', 'JPM': 'JPM',
    'NVDA': 'NVDA', 'PFE': 'PFE', 'TGT': 'TGT', 'XOM': 'XOM'
}

default_start_date = datetime.date(1999, 12, 31)
default_through_date = datetime.date(2025, 12, 31)

def run_distribution_analysis(start_date, through_date):
    if start_date is None or through_date is None:
        print('Select both dates.')
        return
    if start_date >= through_date:
        print('The start date must be earlier than the through-date.')
        return
    end_date = through_date + datetime.timedelta(days=1)  # yfinance end is exclusive

    adj_close_df = pd.DataFrame()
    print(f"Downloading data for: {list(ticker_map.values())} from {start_date} through {through_date}...")

    try:
        data_downloaded = yf.download(
            tickers,
            start=start_date,
            end=end_date,
            interval='1mo',
            group_by='ticker',
            progress=False,
            auto_adjust=True,
            actions=False
        )
        if data_downloaded.empty:
            raise ValueError("yf.download returned an empty DataFrame.")

        # With auto_adjust=True, Close is the adjusted-close series.
        df_list = []
        for ticker in tickers:
            if ticker in data_downloaded.columns.get_level_values(0):
                ticker_data = data_downloaded[ticker]
                if 'Close' in ticker_data.columns and not ticker_data['Close'].isnull().all():
                    df_list.append(ticker_data['Close'].rename(ticker_map[ticker]))

        if not df_list:
            raise ValueError("No valid adjusted-close data extracted for any ticker.")

        adj_close_df = pd.concat(df_list, axis=1)

        # FIX: Drop rows with all NaNs, but don't forward-fill
        # Forward-filling creates artificial price continuity
        adj_close_df = adj_close_df.dropna(how='all')

    except Exception as e:
        print(f"Data Acquisition Error: {e}")
        monthly_returns = pd.DataFrame()

    if not adj_close_df.empty:
        monthly_returns = adj_close_df.pct_change(fill_method=None)

        # FIX: Drop rows with all NaNs first
        monthly_returns = monthly_returns.dropna(how='all')

        # FIX: Filter columns with insufficient data (consistent with optimization codes)
        min_observations = 12  # At least 12 months of data
        monthly_returns = monthly_returns.loc[:, monthly_returns.notna().sum() >= min_observations]

        if monthly_returns.empty:
            print(f"No assets have at least {min_observations} observations.")
            monthly_returns = pd.DataFrame()
        else:
            # FIX: For statistics, we want per-asset analysis, so we keep individual series
            # We do NOT use dropna() for common intersection here, unlike optimization
            # Each asset is analyzed independently with its available data
            pass
    else:
        monthly_returns = pd.DataFrame()

    display_order = ['SP500', 'NDQ100', 'EEM', 'Gold Futures', 'AMD', 'AMZN', 'FORD', 'JPM', 'NVDA', 'PFE', 'TGT', 'XOM']
    available_cols = [col for col in display_order if col in monthly_returns.columns]
    monthly_returns = monthly_returns.reindex(columns=available_cols)

    # --- 2. Generate Histograms (First Slide) ---
    if not monthly_returns.empty:
        print("\n### Risk in Investments")
        print("\n**Histograms of various stock/index returns**")
        print(f"Assets with sufficient data: {', '.join(monthly_returns.columns)}\n")

        num_tickers = len(monthly_returns.columns)
        nrows = (num_tickers + 3) // 4
        fig, axes = plt.subplots(nrows, 4, figsize=(16, 4 * nrows))
        axes = axes.flatten()

        for i, ticker in enumerate(monthly_returns.columns):
            ax = axes[i]
            # Drop NaNs for each asset individually
            returns = monthly_returns[ticker].dropna()

            if len(returns) < 2:
                ax.text(0.5, 0.5, 'Insufficient Data', ha='center', va='center', transform=ax.transAxes)
                ax.set_xlabel(ticker)
                continue

            ax.hist(returns, bins=15, density=True, color='darkkhaki', alpha=0.8)

            mu, std = returns.mean(), returns.std()
            if std > 1e-9:  # Only plot normal curve if std is meaningful
                x = np.linspace(returns.min(), returns.max(), 100)
                p = norm.pdf(x, mu, std)
                ax.plot(x, p, linewidth=1.5, color='steelblue')

            ax.set_xlabel(ticker)
            ax.set_ylabel('Density')
            ax.grid(True, linestyle=':', alpha=0.6)

        for j in range(i + 1, len(axes)):
            axes[j].set_visible(False)

        plt.tight_layout()
        plt.show()

        print("\n---\n")

        # --- 3. Calculate and Display Summary Statistics (Second Slide) ---
        print("### Risk in Investments\n")

        summary_stats = pd.DataFrame(index=[
            'Mean', 'Standard Error', 'Median', 'Standard Deviation', 'Sample Variance',
            'CoV', 'Excess Kurtosis', 'Skewness', 'Range', 'Minimum', 'Maximum', 'Count'
        ])

        for ticker in monthly_returns.columns:
            # Drop NaNs for each asset individually
            returns = monthly_returns[ticker].dropna()

            if len(returns) < 2:
                # Not enough data for statistics
                summary_stats[ticker] = pd.Series({
                    'Mean': np.nan, 'Standard Error': np.nan, 'Median': np.nan,
                    'Standard Deviation': np.nan, 'Sample Variance': np.nan,
                    'CoV': np.nan, 'Excess Kurtosis': np.nan, 'Skewness': np.nan,
                    'Range': np.nan, 'Minimum': np.nan, 'Maximum': np.nan,
                    'Count': len(returns)
                })
                continue

            mean_val = returns.mean()
            std_val = returns.std()

            stats = {
                'Mean': mean_val,
                'Standard Error': returns.sem(),
                'Median': returns.median(),
                'Standard Deviation': std_val,
                'Sample Variance': returns.var(),
                'CoV': std_val / mean_val if abs(mean_val) > 1e-9 else np.nan,
                'Excess Kurtosis': kurtosis(returns, fisher=True, bias=False) if len(returns) >= 4 else np.nan,
                'Skewness': skew(returns, bias=False) if len(returns) >= 3 else np.nan,
                'Range': returns.max() - returns.min(),
                'Minimum': returns.min(),
                'Maximum': returns.max(),
                'Count': returns.count()
            }
            summary_stats[ticker] = pd.Series(stats)

        def format_percent(x):
            return f"{x:.2%}"

        formatters = {
            'Mean': format_percent,
            'Standard Error': format_percent,
            'Median': format_percent,
            'Standard Deviation': format_percent,
            'Minimum': format_percent,
            'Maximum': format_percent,
            'Range': format_percent
        }

        summary_display = summary_stats.style.format(formatters, na_rep="-").format(
            subset=pd.IndexSlice[['Sample Variance', 'CoV', 'Excess Kurtosis', 'Skewness'], :],
            formatter="{:.4f}"
        ).format(
            subset=pd.IndexSlice[['Count'], :],
            formatter="{:.0f}"
        )

        display(summary_display)

    else:
        print("Could not generate output because no valid data was downloaded.")
distribution_start_box = widgets.Text(
    value=default_start_date.isoformat(), description='Start:'
)
distribution_through_box = widgets.Text(
    value=default_through_date.isoformat(), description='Through:'
)
distribution_run_button = widgets.Button(description='Download and run', button_style='primary')
distribution_output = widgets.Output()

def on_distribution_run(_):
    with distribution_output:
        clear_output(wait=True)
        try:
            start_date = datetime.date.fromisoformat(distribution_start_box.value.strip().replace('.', '-'))
            through_date = datetime.date.fromisoformat(distribution_through_box.value.strip().replace('.', '-'))
        except ValueError:
            print('Invalid date. Use YYYY-MM-DD or YYYY.MM.DD.')
            return
        run_distribution_analysis(start_date, through_date)

distribution_run_button.on_click(on_distribution_run)
display(widgets.VBox([
    widgets.HBox([distribution_start_box, distribution_through_box]),
    distribution_run_button,
    distribution_output
]))
```

    *   A positively skewed distribution is a type of data distribution where most of the data points are clustered towards the lower end of the spectrum, with a smaller number of outliers near the higher end.

### 1. Phân phối Không lệch / Đối xứng (Symmetrical / No Skew)

* **Nội dung:** Trong phân phối chuẩn (không bị lệch), hình dáng biểu đồ hoàn toàn đối xứng qua tâm.
* **Đặc điểm:** **Giá trị trung bình (Mean) = Trung vị (Median) = Yếu vị (Mode)**.
* **Ví dụ ngắn:** Điểm thi môn Kinh tế vĩ mô của lớp. Điểm trung bình là 7, điểm nằm chính giữa lớp (trung vị) là 7, và số điểm nhiều người đạt nhất (yếu vị) cũng là 7. Mọi thứ đối xứng hoàn hảo.

---

### 2. Phân phối Lệch Dương / Lệch Phải (Positively Skewed Distribution)

* **Nội dung:** Phần lớn dữ liệu tập trung ở **mức lợi nhuận thấp/trung bình** (phía bên trái), nhưng có một vài ngoại lệ (outliers) kéo cái đuôi dài ra phía **bên phải (lợi nhuận rất cao)**.
* **Đặc điểm:** $\text{Mean} > \text{Median} > \text{Mode}$.
* **Ví dụ ngắn (Vé số / Cổ phiếu Start-up):**
* Hầu như ai mua vé số cũng hòa tiền hoặc lỗ nhẹ (dữ liệu dồn ở bên trái).
* Nhưng có một vài người trúng độc đắc hàng chục tỷ (ngoại lệ cực lớn bên phải).
* Đối với nhà đầu tư, **lệch dương là rủi ro "vui"** vì xác suất lỗ nặng thấp, nhưng lại có cơ hội trúng lớn.



---

### 3. Phân phối Lệch Âm / Lệch Trái (Negatively Skewed Distribution)

* **Nội dung:** Phần lớn dữ liệu tập trung ở **mức lợi nhuận khá/tốt** (phía bên phải), nhưng cái đuôi dài lại bị kéo về **bên trái (các khoản thua lỗ rất nặng)**.
* **Đặc điểm:** $\text{Mean} < \text{Median} < \text{Mode}$.
* **Ví dụ ngắn (Bán bảo hiểm / Cho vay tín dụng):**
* Mỗi tháng bạn thu tiền phí bảo hiểm đều đặn và kiếm lời nhỏ (dữ liệu dồn bên phải).
* Nhưng khi có thiên tai/tai nạn xảy ra, bạn phải bồi thường một số tiền khổng lồ (kéo cái đuôi lỗ nặng sang bên trái).
* Đối với nhà đầu tư, **lệch âm là rủi ro "nguy hiểm"** vì bình thường trông rất ổn, nhưng khi gặp chuyện xui thì cháy tài khoản.



---

### 4. Hệ số Biến động (Coefficient of Variation - CoV)

* **Nội dung:** Đo lường **độ biến động tương đối (rủi ro) so với giá trị trung bình (lợi nhuận)**.
* **Công thức:**

$$\text{CoV} = \frac{\text{Độ lệch chuẩn } (\sigma)}{\text{Lợi nhuận trung bình } (\mu)}$$


* **Ý nghĩa:** Trả lời cho câu hỏi: *"Để kiếm được 1% lợi nhuận, bạn phải gánh chịu bao nhiêu đơn vị rủi ro?"* (Chỉ số CoV càng nhỏ càng tốt).
* **Ví dụ ngắn:**
* **Cổ phiếu A:** Lợi nhuận $10\%$, Độ lệch chuẩn $5\% \rightarrow \text{CoV} = 5/10 = 0.5$.
* **Cổ phiếu B:** Lợi nhuận $20\%$, Độ lệch chuẩn $16\% \rightarrow \text{CoV} = 16/20 = 0.8$.
* **Nhận xét:** Dù cổ phiếu B cho lợi nhuận cao hơn ($20\% > 10\%$), nhưng nếu tính trên mỗi unit lợi nhuận tạo ra, cổ phiếu A an toàn hơn vì có CoV thấp hơn ($0.5 < 0.8$).

```python
import numpy as np

# Separate five-state example: Delta Corp.
probs = [0.15, 0.30, 0.30, 0.20, 0.05]
returns = [0.44, 0.23, 0.06, -0.09, -0.40]
rf = 0.04
P0 = 50
prices = [70, 60, 52, 45, 30]
dividends = [2.00, 1.50, 1.00, 0.50, 0.00]

# Total Returns
print("Alternative example: Delta Corp.")
total_returns = [(p1 - P0 + d) / P0 for p1, d in zip(prices, dividends)]
print("Total Returns:", [f"{r*100:.2f}%" for r in total_returns])

# Expected Return
E_r = sum(p * r for p, r in zip(probs, returns))
print(f"E(r): {E_r * 100:.2f}%")

# Variance
var = sum(p * (r - E_r)**2 for p, r in zip(probs, returns))
print(f"Variance: {var:.6f}")

# Standard Deviation
std = np.sqrt(var)
print(f"Std Dev: {std * 100:.2f}%")

# Excess Returns
excess_returns = [r - rf for r in returns]
print("Excess Returns:", [f"{er*100:.2f}%" for er in excess_returns])

# Expected Excess Return
E_excess = sum(p * er for p, er in zip(probs, excess_returns))
print(f"E(r) - rf: {E_excess * 100:.2f}%")

# Variance of Excess Return
var_excess = sum(p * (er - E_excess)**2 for p, er in zip(probs, excess_returns))
print(f"Excess Variance: {var_excess:.6f}")
```

```python
# If needed, uncomment and run once:
# %pip install -U yfinance curl_cffi pandas numpy matplotlib ipywidgets

import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, HTML, clear_output
import datetime

# --- 1. Data Download and Preparation ---
tickers = ['^SP500TR', 'QQQ', 'EEM', 'GC=F', 'AMD', 'AMZN', 'F', 'JPM', 'NVDA', 'PFE', 'TGT', 'XOM']
ticker_map = {
    '^SP500TR': 'SP500', 'QQQ': 'NDQ100', 'EEM': 'EEM', 'GC=F': 'Gold Futures',
    'AMD': 'AMD', 'AMZN': 'AMZN', 'F': 'FORD', 'JPM': 'JPM',
    'NVDA': 'NVDA', 'PFE': 'PFE', 'TGT': 'TGT', 'XOM': 'XOM'
}

default_start_date = datetime.date(1999, 12, 31)
default_split_date = datetime.date(2010, 3, 31)
last_month_end = (pd.Timestamp.today().normalize().replace(day=1) - pd.Timedelta(days=1)).date()

def run_risk_return_analysis(start_date, through_date, split_date):
    if start_date is None or through_date is None or split_date is None:
        print('Select all three dates.')
        return
    if not start_date < split_date < through_date:
        print('The split date must fall between the start and through dates.')
        return
    end_date = through_date + datetime.timedelta(days=1)  # yfinance end is exclusive

    print(f"Downloading data for specified tickers from {start_date} through {through_date}...")

    try:
        data_downloaded = yf.download(
            tickers, start=start_date, end=end_date, interval='1mo',
            group_by='ticker', progress=False, auto_adjust=True, actions=False
        )
        if data_downloaded.empty:
            raise ValueError("yf.download returned an empty DataFrame.")

        # With auto_adjust=True, Close is the adjusted-close series.
        df_list = []
        for ticker in tickers:
            if ticker in data_downloaded.columns.get_level_values(0):
                ticker_data = data_downloaded[ticker]
                if 'Close' in ticker_data.columns and not ticker_data['Close'].isnull().all():
                    df_list.append(ticker_data['Close'].rename(ticker_map[ticker]))

        if not df_list:
            raise ValueError("No valid adjusted-close data extracted for any ticker.")

        adj_close_df = pd.concat(df_list, axis=1)

        # FIX: Don't forward-fill, just drop rows with all NaNs
        adj_close_df = adj_close_df.dropna(how='all')

    except Exception as e:
        print(f"Data Acquisition Error: {e}")
        adj_close_df = pd.DataFrame()

    if not adj_close_df.empty:
        monthly_returns = adj_close_df.pct_change(fill_method=None)

        # FIX: Drop rows with all NaNs
        monthly_returns = monthly_returns.dropna(how='all')

        # FIX: Filter columns with insufficient data
        min_observations = 12
        monthly_returns = monthly_returns.loc[:, monthly_returns.notna().sum() >= min_observations]

        if monthly_returns.empty:
            print(f"No assets have at least {min_observations} observations.")
            monthly_returns = pd.DataFrame()
        else:
            display_order = ['SP500', 'NDQ100', 'EEM', 'Gold Futures', 'AMD', 'AMZN', 'FORD', 'JPM', 'NVDA', 'PFE', 'TGT', 'XOM']
            available_cols = [col for col in display_order if col in monthly_returns.columns]
            monthly_returns = monthly_returns.reindex(columns=available_cols)

            print(f"Assets with sufficient data: {', '.join(monthly_returns.columns)}\n")
    else:
        monthly_returns = pd.DataFrame()

    # --- 2. Define Sample Periods ---
    if not monthly_returns.empty:
        full_sample_start = datetime.datetime.combine(start_date, datetime.time.min)
        full_sample_end = datetime.datetime.combine(through_date, datetime.time.max)
        sub_sample_1_start = full_sample_start
        sub_sample_1_end = datetime.datetime.combine(split_date, datetime.time.max)
        sub_sample_2_start = datetime.datetime.combine(
            split_date + datetime.timedelta(days=1), datetime.time.min
        )
        sub_sample_2_end = datetime.datetime.combine(through_date, datetime.time.max)

        # Slice the data according to the defined periods
        full_sample_data = monthly_returns.loc[full_sample_start:full_sample_end]
        sub_sample_1_data = monthly_returns.loc[sub_sample_1_start:sub_sample_1_end]
        sub_sample_2_data = monthly_returns.loc[sub_sample_2_start:sub_sample_2_end]

        # FIX: Check if subsamples are empty
        samples = {
            'Full Sample': (full_sample_data, full_sample_start, full_sample_end),
            'Sub-sample 1': (sub_sample_1_data, sub_sample_1_start, sub_sample_1_end),
            'Sub-sample 2': (sub_sample_2_data, sub_sample_2_start, sub_sample_2_end)
        }

        for name, (data, start, end) in samples.items():
            if data.empty:
                print(f"Warning: {name} ({start:%Y-%m-%d} to {end:%Y-%m-%d}) has no data.")

    # --- 3. Define Plotting Function ---
    def create_risk_return_plot(ax, data, title):
        if data.empty:
            ax.text(0.5, 0.5, 'No Data Available', ha='center', va='center',
                    transform=ax.transAxes, fontsize=14)
            ax.set_title(title, fontsize=12)
            return

        # FIX: Drop NaNs per column for mean/std calculation
        # For risk-return, we analyze each asset independently
        means = data.mean()
        stds = data.std()

        # Filter out assets with insufficient data in this period
        valid_assets = means.notna() & stds.notna() & (stds > 0)
        means = means[valid_assets]
        stds = stds[valid_assets]

        if means.empty:
            ax.text(0.5, 0.5, 'Insufficient Data', ha='center', va='center',
                    transform=ax.transAxes, fontsize=14)
            ax.set_title(title, fontsize=12)
            return

        risk_return = pd.DataFrame({'Return': means, 'Risk': stds})
        ax.scatter(risk_return['Risk'], risk_return['Return'], color='#004B87', s=50)

        for ticker, row in risk_return.iterrows():
            ax.text(row['Risk'] * 1.02, row['Return'], ticker, fontsize=10,
                    verticalalignment='center')

        ax.set_xlabel('Risk (Standard Deviation)', fontsize=10)
        ax.set_ylabel('Expected Return (Mean)', fontsize=10)
        ax.set_title(title, fontsize=12)
        ax.grid(True, linestyle='-', alpha=0.3)
        ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{x:.2%}'))
        ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{y:.2%}'))

    # --- 4. Generate the Three Plots ---
    if not monthly_returns.empty:
        # --- Plot 1: Full Sample ---
        fig1, ax1 = plt.subplots(1, 1, figsize=(10, 6))
        title1 = (
            f"Risk & Return (Monthly, {start_date:%Y-%m-%d} to "
            f"{through_date:%Y-%m-%d})"
        )
        create_risk_return_plot(ax1, full_sample_data, title1)
        plt.tight_layout()
        plt.show()

        # --- Plot 2 & 3: Sub-samples ---
        fig2, (ax2, ax3) = plt.subplots(1, 2, figsize=(18, 7))

        title2 = (
            f"Risk & Return (Monthly, {start_date:%Y-%m-%d} to "
            f"{split_date:%Y-%m-%d})"
        )
        create_risk_return_plot(ax2, sub_sample_1_data, title2)

        second_period_start = split_date + datetime.timedelta(days=1)
        title3 = (
            f"Risk & Return (Monthly, {second_period_start:%Y-%m-%d} to "
            f"{through_date:%Y-%m-%d})"
        )
        create_risk_return_plot(ax3, sub_sample_2_data, title3)

        plt.tight_layout()
        plt.show()
    else:
        print("Could not generate plots because no valid data was downloaded.")
risk_return_start_box = widgets.Text(
    value=default_start_date.isoformat(), description='Start:'
)
risk_return_through_box = widgets.Text(
    value=last_month_end.isoformat(), description='Through:'
)
risk_return_split_box = widgets.Text(
    value=default_split_date.isoformat(), description='Split:'
)
risk_return_run_button = widgets.Button(description='Download and run', button_style='primary')
risk_return_output = widgets.Output()

def on_risk_return_run(_):
    with risk_return_output:
        clear_output(wait=True)
        try:
            start_date = datetime.date.fromisoformat(risk_return_start_box.value.strip().replace('.', '-'))
            through_date = datetime.date.fromisoformat(risk_return_through_box.value.strip().replace('.', '-'))
            split_date = datetime.date.fromisoformat(risk_return_split_box.value.strip().replace('.', '-'))
        except ValueError:
            print('Invalid date. Use YYYY-MM-DD or YYYY.MM.DD.')
            return
        run_risk_return_analysis(start_date, through_date, split_date)

risk_return_run_button.on_click(on_risk_return_run)
display(widgets.VBox([
    widgets.HBox([risk_return_start_box, risk_return_split_box, risk_return_through_box]),
    risk_return_run_button,
    risk_return_output
]))
```

### Risk in Investments
*   True means and variances are unobservable
    *   Possible scenarios like the ones used in the examples are unknown
*   Means and variances must therefore be estimated from historical observations

*   Expected Return Across States
    
    $$E(r)=\sum_{s=1}^{S}p(s)r(s)$$

*   Historical Arithmetic Average
    
    $$\bar r=\frac{1}{n}\sum_{t=1}^{n}r_t$$
    
    *   This is the equally weighted sample counterpart of expected return
*   Geometric (Time-Weighted) Average
    *   Terminal value of the investment:
        $$TV_n=(1+r_1)(1+r_2)\cdots(1+r_n)$$
    *   Geometric average:
        $$g=TV_n^{1/n}-1$$

---




### 1. Vấn đề cốt lõi: Không ai biết trước tương lai

* **Thực tế:** Giá trị trung bình ($\mu$) và phương sai ($\sigma^2$) thực sự của tài sản là **không thể quan sát trực tiếp** (unobservable).
* Các kịch bản kinh tế trong tương lai (bùng nổ, suy thoái...) chỉ là dự đoán chứ không biết chắc chắn.
* **Giải pháp:** Do đó, nhà đầu tư phải **ước lượng (estimate)** lợi nhuận và rủi ro dựa trên **dữ liệu lịch sử (historical observations)**.

---

### 2. Ba công thức tính Lợi nhuận trong Slide

#### ① Lợi nhuận kỳ vọng theo kịch bản (Expected Return Across States)

* **Ý nghĩa:** Dùng khi bạn đưa ra các **kịch bản tương lai** và gán cho mỗi kịch bản một **xác suất** xảy ra.
* **Công thức:**

$$E(r) = \sum_{s=1}^{S} p(s) r(s)$$


* **Ví dụ ngắn:** Bạn dự đoán năm tới có $60\%$ khả năng kinh tế tốt (lời $20\%$) và $40\%$ khả năng kinh tế xấu (lỗ $10\%$).

$$\text{Lợi nhuận kỳ vọng } E(r) = (0.6 \times 20\%) + (0.4 \times -10\%) = 12\% - 4\% = 8\%$$



---

#### ② Lợi nhuận trung bình đại số (Historical Arithmetic Average)

* **Ý nghĩa:** Lấy tổng lợi nhuận của các kỳ trong quá khứ rồi chia đều cho số kỳ (coi trọng số các kỳ là như nhau - equally weighted).
* **Công thức:**

$$\bar{r} = \frac{1}{n} \sum_{t=1}^{n} r_t$$


* **Ví dụ ngắn:** Giá cổ phiếu 3 năm qua lần lượt tăng $10\%$, $20\%$, và $-6\%$.

$$\text{Trung bình đại số } \bar{r} = \frac{10\% + 20\% - 6\%}{3} = \frac{24\%}{3} = 8\%/\text{năm}$$



---

#### ③ Lợi nhuận trung bình hình học (Geometric / Time-Weighted Average)

* **Ý nghĩa:** Tính đến **hiệu ứng lãi kép** qua các năm để đo lường mức tăng trưởng **thực tế** của tài sản từ đầu đến cuối kỳ.
* **Công thức:**
* *Giá trị cuối cùng của khoản đầu tư ($TV_n$):*

$$TV_n = (1 + r_1)(1 + r_2)\dots(1 + r_n)$$


* *Lợi nhuận trung bình hình học ($g$):*

$$g = TV_n^{1/n} - 1$$




* **Ví dụ ngắn:** Bạn có $100$ USD. Năm 1 bạn **lời $100\%$** (lên $200$ USD). Năm 2 bạn **lỗ $50\%$** (về lại $100$ USD).
* *Nếu tính trung bình đại số:* $\frac{100\% + (-50\%)}{2} = 25\%/\text{năm}$ (Trông có vẻ lời, nhưng thực tế bạn $0$ đồng tiền lời!).
* *Tính trung bình hình học ($g$):*

$$TV_2 = (1 + 1.00) \times (1 - 0.50) = 2 \times 0.5 = 1$$


$$g = 1^{1/2} - 1 = 0\%$$




> **Bài học:** Trung bình hình học $g = 0\%$ phản ánh **chính xác thực tế** là bạn không hề kiếm được đồng nào sau 2 năm!



---

### 💡 Tóm lại cho sinh viên dễ nhớ

* **Expected Return:** Dùng để **dự báo tương lai** (khi có xác suất $p(s)$).
* **Arithmetic Average ($\bar{r}$):** Dùng để **dự đoán lợi nhuận của 1 năm bất kỳ** trong tương lai dựa trên quá khứ.
* **Geometric Average ($g$):** Dùng để **đo lường hiệu suất thực tế** của cả giai đoạn đầu tư nhiều năm (có tính lãi kép).

### Risk in Investments
*   Sample Variance
    *   An unbiased estimator of the return variance under the usual assumptions:
        $$s^2 = \frac{1}{n-1}\sum_{t=1}^{n}[r_t-\bar r]^2$$
*   Sample Standard Deviation
    $$s=\sqrt{s^2}$$
*   Sharpe Ratio
    $$\frac{\text{Risk premium}}{\text{SD of excess returns}}$$

### 1. Phương sai mẫu (Sample Variance - $s^2$)

* **Ý nghĩa:** Do chúng ta không biết trước tương lai hay toàn bộ quần thể dữ liệu, ta dùng dữ liệu lịch sử của $n$ kỳ để ước lượng độ biến động của lợi nhuận. Đây là một **ước lượng không chệch (unbiased estimator)** cho phương sai thực tế.
* **Công thức:**

$$s^2 = \frac{1}{n-1}\sum_{t=1}^{n}[r_t-\bar r]^2$$



*(Lưu ý: Mẫu số là $n-1$ thay vì $n$ để điều chỉnh sai lệch khi tính toán trên tập dữ liệu mẫu).*
* **Ví dụ ngắn:** Bạn theo dõi lợi nhuận cổ phiếu A trong 3 năm qua lần lượt là $10\%$, $20\%$, $-6\%$ (Trung bình $\bar r = 8\%$).
* Tính độ lệch so với trung bình của từng năm rồi bình phương:
* Năm 1: $(10\% - 8\%)^2 = 2\%^2 = 0.0004$
* Năm 2: $(20\% - 8\%)^2 = 12\%^2 = 0.0144$
* Năm 3: $(-6\% - 8\%)^2 = (-14\%)^2 = 0.0196$


* Cộng lại rồi chia cho $(3 - 1) = 2$:

$$s^2 = \frac{0.0004 + 0.0144 + 0.0196}{2} = 0.0172$$





---

### 2. Độ lệch chuẩn mẫu (Sample Standard Deviation - $s$)

* **Ý nghĩa:** Phương sai $s^2$ ở trên có đơn vị là "phần trăm bình phương" ($\%^2$), rất khó dùng để hình dung. Do đó, ta lấy căn bậc hai của phương sai để đưa thước đo rủi ro về **cùng đơn vị tính (%) với lợi nhuận**.
* **Công thức:**

$$s = \sqrt{s^2}$$


* **Ví dụ tiếp tục từ mục 1:**

$$s = \sqrt{0.0172} \approx 0.1311 = 13.11\%$$


> **Ý nghĩa:** Cổ phiếu A có rủi ro/độ biến động trung bình quanh mức trung bình là $13.11\%$.



---

### 3. Tỷ số Sharpe (Sharpe Ratio)

* **Ý nghĩa:** Đây là một trong những chỉ số quan trọng nhất trong quản lý danh mục đầu tư. Sharpe Ratio đo lường **phần lợi nhuận vượt mức (Risk Premium) nhận được trên mỗi đơn vị rủi ro tổng thể (SD)**.
* **Công thức:**

$$\text{Sharpe Ratio} = \frac{\text{Risk premium}}{\text{SD of excess returns}} = \frac{E(r) - r_f}{s}$$


* $\text{Risk premium} = E(r) - r_f$: Lợi nhuận kỳ vọng của cổ phiếu trừ đi lãi suất phi rủi ro (gửi tiết kiệm ngân hàng).
* $\text{SD of excess returns} = s$: Độ lệch chuẩn của lợi nhuận.


* **Chỉ số Sharpe càng cao $\rightarrow$ Danh mục đầu tư càng hiệu quả!**

#### Ví dụ ngắn dễ hiểu cho Sinh viên:

Giả sử lãi suất gửi tiết kiệm ngân hàng an toàn ($r_f$) là **$4\%$**:

* **Cổ phiếu A:** Lợi nhuận $10\%$, Độ lệch chuẩn ($s$) = $12\%$

$$\text{Sharpe Ratio (A)} = \frac{10\% - 4\%}{12\%} = \frac{6\%}{12\%} = 0.5$$


* **Cổ phiếu B:** Lợi nhuận $16\%$, Độ lệch chuẩn ($s$) = $30\%$

$$\text{Sharpe Ratio (B)} = \frac{16\% - 4\%}{30\%} = \frac{12\%}{30\%} = 0.4$$



> **Nhận xét:** Dù cổ phiếu B cho lợi nhuận cao hơn ($16\% > 10\%$), nhưng cổ phiếu A lại có **Sharpe Ratio cao hơn ($0.5 > 0.4$)**. Điều này có nghĩa là cổ phiếu A mang lại phần tiền lời "đền bù" tốt hơn cho mỗi $1\%$ rủi ro mà bạn phải gánh chịu!

### Risk in Investments
*   Symmetric Returns
    *   Standard deviation is a good measure of risk
    *   If returns are jointly normal, linear portfolio returns are also symmetric
    *   Under jointly normal returns, the mean vector and covariance matrix fully characterize portfolio returns
    *   Under elliptical/normal returns, the correlation matrix summarizes dependence; otherwise correlations can miss tail dependence or asymmetry
*   What if excess returns are not normally distributed?
    *   STD is no longer a complete measure of risk
    *   Skewness:
        $$
        \text{Skew} = \text{Average} \left[\frac{(R - \bar{R})^3}{\hat{\sigma}^3}\right]
        $$
    
    *   Excess kurtosis:
        $$
        \text{Kurtosis} = \text{Average} \left[\frac{(R - \bar{R})^4}{\hat{\sigma}^4}\right] - 3
        $$
    *   Sharpe ratio is not a complete measure of portfolio performance

### 1. Khi lợi nhuận đối xứng (Symmetric Returns)

* **Trường hợp lý tưởng:** Nếu phân phối lợi nhuận đối xứng (như Phân phối chuẩn), **Độ lệch chuẩn ($\sigma$) là thước đo rủi ro hoàn hảo**.
* **Tính chất danh mục:** Khi lợi nhuận các tài sản tuân theo phân phối chuẩn đồng thời (jointly normal), lợi nhuận của cả danh mục kết hợp cũng sẽ đối xứng. Lúc này, chỉ cần **Lợi nhuận trung bình (Mean)** và **Ma trận phương sai/hiệu phương sai (Covariance Matrix)** là đủ để mô tả trọn vẹn rủi ro và lợi nhuận của cả danh mục.
* **Ma trận tương quan (Correlation matrix):** Giúp tóm tắt mối quan hệ giữa các tài sản.

### 2. Nếu lợi nhuận VƯỢT MỨC KHÔNG tuân theo Phân phối chuẩn?

Trong thực tế thị trường tài chính, lợi nhuận cổ phiếu/tiền mã hóa thường **bị lệch** và có **nhiều sự kiện sụp đổ bất ngờ**. Khi đó:

* **Độ lệch chuẩn ($\sigma$) không còn đo lường đầy đủ rủi ro nữa.** Nó bị "mù" trước các kịch bản thua lỗ cực nặng ở vùng đuôi.
* Ta phải dùng thêm 2 chỉ số thống kê bậc cao:

#### ① Độ lệch (Skewness - Đo tính bất đối xứng)

* **Công thức:**

$$\text{Skew} = \text{Average} \left[\frac{(R - \bar{R})^3}{\hat{\sigma}^3}\right]$$


* **Ý nghĩa:** Đo xem đuôi phân phối bị kéo dài về bên trái (lỗ nặng) hay bên phải (lãi lớn). Nếu Skewness $< 0$, bạn có nguy cơ gặp các vụ sập giá bất ngờ.

#### ② Độ nhọn vượt mức (Excess Kurtosis - Đo độ "dày" của đuôi)

* **Công thức:**

$$\text{Kurtosis} = \text{Average} \left[\frac{(R - \bar{R})^4}{\hat{\sigma}^4}\right] - 3$$


* **Ý nghĩa:** So sánh với phân phối chuẩn (có Kurtosis = 3, nên Excess Kurtosis = 0).
* Nếu Excess Kurtosis $> 0$ (đuôi dày / fat tails): Tín hiệu cho thấy các **sự kiện thiên nga đen (sập sàn hoặc tăng vọt cực đoan)** xuất hiện nhiều hơn hẳn so với lý thuyết thông thường.



#### ③ Tỷ số Sharpe không còn là thước đo hiệu quả hoàn hảo

* Vì Sharpe Ratio dùng Độ lệch chuẩn ($\sigma$) ở mẫu số, nên khi thị trường có Skewness âm nặng hoặc Kurtosis cao, **Sharpe Ratio sẽ đánh giá thấp rủi ro thực tế**, làm cho danh mục trông có vẻ "an toàn" hơn so với bản chất thực sự của nó.

Ví dụ về Skewness (Độ lệch)

Std chỉ đo "trung bình độ phân tán" — KHÔNG đo "hình dạng" phân phối
Công thức Std:

σ = √("Average" [(R − bar(R))^2 ])

Để ý: công thức này chỉ lấy trung bình của bình phương độ lệch. Nó không quan tâm:

	* Độ lệch đó đến từ nhiều lần lệch nhỏ hay vài lần lệch cực lớn
	* Độ lệch đó nằm bên trái (lỗ) hay bên phải (lãi)

Hai chuỗi dữ liệu hoàn toàn khác nhau về "hình dạng" vẫn có thể cho ra cùng một con số Std, miễn là tổng bình phương độ lệch bằng nhau.

Ví dụ minh họa bằng số cụ thể
Tài sản X (biến động đều):
5 ngày lợi nhuận: -4%, -2%, 0%, +2%, +4%

	* Mean = 0%
	* Độ lệch mỗi ngày: -4, -2, 0, +2, +4
	* Bình phương: 16, 4, 0, 4, 16 → Tổng = 40 → Trung bình = 8
	* Std = √8 ≈ 2.83%

Tài sản Y (hầu hết yên ổn, 1 ngày sập mạnh):
5 ngày lợi nhuận: +0.5%, +0.5%, +0.5%, +0.5%, -15%

	* Mean = (0.5+0.5+0.5+0.5-15)/5 = -2.6%
	* Độ lệch so với mean: 3.1, 3.1, 3.1, 3.1, -12.4
	* Bình phương: 9.6, 9.6, 9.6, 9.6, 153.8 → Tổng ≈ 192 → Trung bình ≈ 38.4
	* Std = √38.4 ≈ 6.2%

Ở ví dụ này Std của Y thực ra cao hơn X — đúng như bạn nghĩ, một cú sập lớn sẽ kéo Std lên cao.

Nhưng vấn đề nằm ở chỗ: nếu Y có NHIỀU ngày yên ổn hơn để "pha loãng" biến động, Std tính trung bình trên toàn bộ mẫu lớn có thể bị kéo xuống thấp, trong khi phân phối thực tế vẫn cực kỳ nguy hiểm. Hãy xem ví dụ thực tế hơn:

Tài sản Y (360 ngày yên ổn, 1 ngày sập):

	* 360 ngày: +0.1% mỗi ngày
	* 1 ngày: -50%

Vì Std được tính trên cả 361 ngày, và 360/361 ngày gần như không đóng góp gì vào phương sai, nên:

σ ≈ √((360 ×(0.1)^2 + 1 ×(50)^2)/361) ≈ √(2500/361) ≈ 2.6%

→ Std chỉ khoảng 2.6%/ngày — nghe có vẻ "an toàn" như một cổ phiếu bình thường! Nhưng thực tế tài sản này chứa rủi ro sập 50% chỉ trong 1 ngày, điều mà Std không "cảnh báo" rõ ràng vì nó bị pha loãng bởi hàng trăm ngày yên ổn.

Vậy tại sao biểu đồ "nhảy đều" trong đầu bạn lại sai?
Bạn đang hình dung Std giống như biên độ dao động cố định (kiểu "mỗi ngày nhảy đúng ±2%"). Nhưng thực tế Std là một con số thống kê tổng hợp từ rất nhiều điểm dữ liệu khác nhau — nó giống "độ rộng trung bình" chứ không phải "biên độ nhảy mỗi lần".

Cùng một Std, bạn có thể đạt được bằng:

	1. Nhiều lần nhảy nhỏ đều đặn (phân phối chuẩn, Kurtosis thấp)
	2. Rất ít lần nhảy nhưng cực lớn + nhiều lần gần như đứng yên (Kurtosis cao, đuôi dày)

Ví dụ hình ảnh dễ nhớ
Hãy tưởng tượng 2 người đi bộ:

Người A: mỗi bước đều dài 2 bước chân, đi đều đặn — rất dễ đoán.

Người B: 99 bước đi rất ngắn (gần như đứng yên), nhưng bước thứ 100 là một cú nhảy xa bằng 20 bước chân.

Nếu tính "độ dài bước trung bình" (kiểu Std) của cả hai, con số có thể ra gần bằng nhau. Nhưng rõ ràng hành vi của Người B nguy hiểm và khó đoán hơn nhiều — đó chính là điều Kurtosis giúp bạn phát hiện ra mà Std bỏ sót.

Đây chính là lý do tại sao trong slide ghi "Std is no longer a complete measure of risk" — nó đo được "độ rộng trung bình" nhưng mù trước "hình dạng" (có đuôi dày hay lệch hay không) của phân phối.


### Risk in Investments

*   The normal distribution is a useful benchmark, but many assets and strategies exhibit skewness and fat tails

*   Skewness and kurtosis
    *   Negative skewness indicates a longer or heavier left tail
    *   Positive excess kurtosis indicates more extreme outcomes than under a normal distribution

*   Implications for VaR and Expected Shortfall (ES/CVaR)
    *   Under normality, mean and standard deviation determine both VaR and ES
    *   Negative skewness and positive excess kurtosis can make normal-based VaR and ES understate downside risk
    *   ES is more sensitive to fat tails because it reflects the magnitude of losses beyond the VaR cutoff

1. Ý chính đầu tiên: "The normal distribution is a useful benchmark, but many assets and strategies exhibit skewness and fat tails"
Đây là lời "thú nhận" của giới tài chính: Mô hình chuẩn (Normal distribution) — nơi mà chỉ cần biết Mean và Std là mô tả được toàn bộ rủi ro — chỉ là một điểm khởi đầu lý thuyết, không phải sự thật.

Thực tế, nhiều tài sản (đặc biệt là cổ phiếu, quỹ hedge fund, chiến lược bán quyền chọn...) có phân phối lợi nhuận bị lệch (skewed) và có đuôi dày (fat tails) — đúng như ví dụ "Tài sản Y sập 50%" mà bạn vừa thấy ở trên.

2. Skewness và Kurtosis — hai "con số bổ sung" cho Std
Skewness (độ lệch)
	* Negative skewness (lệch âm) → đuôi bên trái dài hơn → nghĩa là: thỉnh thoảng có cú lỗ rất lớn, nhưng phần lớn thời gian lãi nhỏ đều đặn.

		* Đây chính xác là trường hợp Tài sản Y trong ví dụ: 360 ngày lãi nhẹ +0.1%, rồi 1 ngày sập -50%.
	* Positive skewness (lệch dương) ngược lại: thỉnh thoảng có cú lãi đột biến lớn, còn lại thường lỗ nhỏ.

Kurtosis (độ nhọn / đuôi dày)
	* Positive excess kurtosis → phân phối có đuôi dày hơn bình thường → xác suất xảy ra sự kiện cực đoan (cực lớn hoặc cực nhỏ) cao hơn so với phân phối chuẩn dự đoán.
	* Đây chính là khái niệm "Người B nhảy 1 bước = 20 bước chân" mà mình vừa ví dụ — Kurtosis cao đo được điều mà Std bỏ sót.

3. Hệ quả đối với VaR và Expected Shortfall (ES/CVaR)
Đây là phần quan trọng nhất về mặt ứng dụng thực tế:

	* VaR (Value at Risk) và ES (Expected Shortfall, hay CVaR) là hai thước đo rủi ro phổ biến dùng để trả lời câu hỏi: "Trong kịch bản xấu, tôi có thể mất bao nhiêu?"
	* Nếu giả định phân phối chuẩn (normal), thì chỉ cần Mean và Std là tính được VaR và ES — y hệt logic Utility Function U = E(r) − ½Aσ² ở slide dưới, nơi chỉ Std đại diện cho rủi ro.

	* Nhưng nếu tài sản thực có skew âm và kurtosis dương (đuôi dày bên trái) như ví dụ Tài sản Y, thì:


		* VaR tính theo công thức chuẩn sẽ đánh giá thấp mức lỗ thực sự có thể xảy ra (vì mô hình chuẩn "không biết" có cú sập 50% đang rình rập).
		* Đây gọi là hiện tượng "underestimate downside risk" — ước lượng rủi ro thấp hơn thực tế, cực kỳ nguy hiểm trong quản trị rủi ro.
	* ES nhạy cảm hơn với đuôi dày (fat tails) so với VaR, vì:


		* VaR chỉ trả lời "ngưỡng lỗ tại mức xác suất X%" — nó là một điểm cắt, không quan tâm bên trong đuôi lỗ nặng đến mức nào.
		* ES thì lấy trung bình của tất cả các khoản lỗ vượt ngưỡng VaR đó → nên nếu đuôi dày (tức có khả năng lỗ cực lớn), ES sẽ phản ánh rõ mức độ nghiêm trọng, còn VaR thì "mù" trước điều này.

4. Kết nối lại với câu hỏi trước của bạn
Slide này chính là lời giải thích "chính thức" cho thắc mắc của bạn ở trên:

Std chỉ là một con số trung bình hóa độ phân tán. Khi dữ liệu có skewness và kurtosis bất thường (giống Tài sản Y), Std sẽ không còn là "thước đo đầy đủ" của rủi ro nữa — đó là lý do tại sao trong thực hành quản trị rủi ro hiện đại, người ta phải bổ sung thêm Skewness, Kurtosis, VaR và ES bên cạnh Std, chứ không chỉ dùng mỗi Std và Mean như trong công thức Utility cổ điển.

    *   Taking a considerable risk for a commensurate gain
    *   Parties have heterogeneous expectations
*   Gambling
    *   Bet on an uncertain outcome for enjoyment

*   Utility Values

    *   Portfolio attractiveness

1. Speculation (Đầu cơ) vs Gambling (Cờ bạc)
Đây là điểm rất hay vì nhiều người hay nhầm hai khái niệm này.

Speculation (Đầu cơ):

	* Chấp nhận rủi ro đáng kể để đổi lấy khoản lợi nhuận tương xứng ("commensurate gain")
	* Quan trọng: các bên có kỳ vọng khác nhau (heterogeneous expectations) — nghĩa là người mua tin giá sẽ lên, người bán tin giá sẽ xuống (hoặc cần thanh khoản/phòng ngừa rủi ro). Mỗi bên có thông tin/niềm tin riêng.
	* Ví dụ: Mua cổ phiếu vì bạn tin công ty sẽ tăng trưởng, trong khi người bán nghĩ khác.

Gambling (Cờ bạc):

	* Đặt cược vào một kết quả không chắc chắn chỉ vì sự giải trí (enjoyment) — không phải vì kỳ vọng lợi nhuận kinh tế thực sự.
	* Quan trọng: các bên gán cùng một xác suất (same probabilities) cho các kết quả có thể xảy ra. Ví dụ: tung đồng xu, ai cũng biết xác suất là 50-50, không ai có "edge" thông tin hơn ai.
	* Về bản chất, đây là một trò chơi tổng-bằng-không (zero-sum), không tạo ra giá trị kinh tế mới.

→ Điểm khác biệt cốt lõi: Đầu cơ dựa trên sự khác biệt về thông tin/kỳ vọng và nhằm mục đích lợi nhuận; Cờ bạc dựa trên xác suất đã biết và chia sẻ, nhằm mục đích giải trí.

2. Utility Values (Giá trị thỏa dụng)
Phần này giới thiệu khung tư duy mà nhà đầu tư sử dụng:

	* Nhà đầu tư sẵn sàng cân nhắc:


		* Tài sản phi rủi ro (risk-free assets) — ví dụ trái phiếu chính phủ
		* Vị thế đầu cơ có phần bù rủi ro dương (positive risk premiums) — nghĩa là chỉ chấp nhận rủi ro nếu được trả thêm lợi nhuận kỳ vọng tương xứng
	* Sự hấp dẫn của danh mục đầu tư (Portfolio attractiveness):


		* Tăng khi expected return tăng (điều hiển nhiên — ai cũng thích lợi nhuận kỳ vọng cao hơn)
		* Giảm khi risk tăng (vì đa số nhà đầu tư là risk-averse — ngại rủi ro)

3. Câu hỏi mở cuối slide: "What happens when return increases with risk?"
Đây là câu hỏi dẫn dắt sang slide tiếp theo (Utility Function U = E(r) − ½Aσ²). Ý tưởng là:

	* Nếu return tăng tỷ lệ thuận với risk, thì có một sự đánh đổi (trade-off) cần được định lượng.
	* Để so sánh các danh mục có return và risk khác nhau, nhà đầu tư cần một thước đo tổng hợp kết hợp cả hai yếu tố — đó chính là Utility Function, sẽ xuất hiện ở slide kế tiếp.
	* Nói cách khác: không thể chỉ nhìn return để chọn danh mục (vì rủi ro cao hơn), cũng không thể chỉ nhìn risk (vì sẽ bỏ lỡ lợi nhuận) — cần một hàm số cân bằng cả hai, tùy theo mức độ "ngại rủi ro" (risk aversion) của từng nhà đầu tư.

### Risk and Risk Aversion
*   Utility Function
    $$U = E(r) - \frac{1}{2}A\sigma^2$$
    *   U = Utility
    *   E(r) = Expected return on the asset or portfolio
    *   A = Coefficient of risk aversion
    *   $\sigma^2$ = Variance of returns
    *   $1/2$ = A scaling factor
*   Risk Averse Investors: $\qquad A > 0$

*   Risk-Neutral Investors: $\quad \; \, A = 0$

*   Risk Lovers: $\qquad \qquad \quad \; A < 0$

1. Ý nghĩa của công thức
U = E(r)−1/2Aσ^2
Hãy hiểu công thức này như một "điểm số hấp dẫn" (attractiveness score) mà mỗi nhà đầu tư gán cho một danh mục/tài sản. Nó có 2 phần đối lập nhau:

	* Phần cộng: E(r) → Lợi nhuận kỳ vọng càng cao, điểm hấp dẫn U càng cao. Đúng với ý "Portfolio attractiveness increases with expected return" ở slide trước.

	* Phần trừ: ½Aσ² → Đây là "hình phạt" (penalty) cho rủi ro. σ² (variance) càng lớn, bị trừ điểm càng nhiều → đúng với ý "decreases with risk".


→ Nói cách khác: U = Lợi nhuận kỳ vọng − Khoản "giá phải trả" cho rủi ro

2. Giải nghĩa từng thành phần
| Ký hiệu | Ý nghĩa |
|---|---|
| **U** | Điểm thỏa dụng (utility) — con số tổng hợp dùng để **so sánh** các danh mục khác nhau. U càng cao → danh mục càng hấp dẫn với nhà đầu tư đó |
| **E(r)** | Lợi nhuận kỳ vọng của tài sản/danh mục |
| **A** | Hệ số ngại rủi ro (risk aversion coefficient) — đặc trưng **cho từng nhà đầu tư**, không phải đặc trưng của tài sản |
| **σ²** | Phương sai (variance) của lợi nhuận — đo độ rủi ro |
| **½** | Hệ số tỷ lệ (scaling factor) — chỉ là quy ước toán học cho công thức "đẹp" hơn khi lấy đạo hàm (không có ý nghĩa kinh tế sâu xa, bạn không cần lo lắng nhiều về nó) |

Điểm mấu chốt cần nhớ: A là "khẩu vị rủi ro" của NHÀ ĐẦU TƯ, không phải của TÀI SẢN
Cùng một danh mục Gamma (risk cao) nhưng:


	* Với nhà đầu tư có A lớn (rất ngại rủi ro) → U sẽ thấp vì bị trừ nhiều

	* Với nhà đầu tư có A nhỏ (ít ngại rủi ro) → U sẽ cao hơn vì bị trừ ít

→ Đây chính là lý do tại sao không có danh mục nào "tốt nhất" cho tất cả mọi người — nó phụ thuộc vào A của từng cá nhân. Điều này bạn sẽ thấy rõ ở ví dụ bảng Alpha/Beta/Gamma trong slide kế tiếp.

3. Ba loại nhà đầu tư dựa theo dấu của A
A > 0: Risk Averse (Ngại rủi ro)

  * Đa số nhà đầu tư thuộc nhóm này.

	* Họ chỉ chấp nhận rủi ro nếu được đền bù xứng đáng bằng lợi nhuận kỳ vọng cao hơn (risk premium dương) — đúng như ý "speculative positions with positive risk premiums" ở slide trước.

	* σ² càng cao → U càng bị kéo xuống càng nhiều → họ "ghét" sự bất định.

A = 0: Risk-Neutral (Trung lập với rủi ro)

  * Phần ½Aσ² = 0 → U = E(r) luôn luôn.

  * Nhóm này chỉ quan tâm lợi nhuận kỳ vọng, hoàn toàn không quan tâm mức độ rủi ro/phương sai.

  * Họ sẽ chọn danh mục có E(r) cao nhất, bất kể σ² lớn tới đâu.

A < 0: Risk Lovers (Thích rủi ro)

  * Phần ½Aσ² trở thành số âm, nên khi trừ đi một số âm = cộng thêm vào U.

  * Nghĩa là: σ² càng cao → U càng tăng → họ thực sự "thích" sự bất định, giống hành vi khi đi đánh bạc (liên hệ lại với khái niệm Gambling ở slide trước!).

  * Đây là nhóm thiểu số, hành vi gần giống người chơi cờ bạc/xổ số hơn là nhà đầu tư thông thường.

Hình dung trực quan
U = E(r) − ½Aσ²

A > 0  (Risk Averse)   : rủi ro → bị PHẠT      → U giảm khi σ² tăng

A = 0  (Risk-Neutral)  : rủi ro → KHÔNG ảnh hưởng → U chỉ phụ thuộc E(r)

A < 0  (Risk Lovers)   : rủi ro → được THƯỞNG    → U tăng khi σ² tăng



### Risk and Risk Aversion

$$

$$

Assume the risk-free rate is 3% and an investor chooses among three portfolios:

| Portfolio | Risk premium | $E(r)$ | $\sigma$ |
| :--- | ---: | ---: | ---: |
| Alpha (low risk) | 3% | 6% | 8% |
| Beta (medium risk) | 7% | 10% | 15% |
| Gamma (high risk) | 12% | 15% | 25% |

| Risk aversion $A$ | $U_{\text{Alpha}}$ | $U_{\text{Beta}}$ | $U_{\text{Gamma}}$ | Highest utility |
| ---: | ---: | ---: | ---: | :--- |
| 2 | $0.06-\frac{1}{2}(2)(0.08^2)=0.0536$ | $0.10-\frac{1}{2}(2)(0.15^2)=0.0775$ | $0.15-\frac{1}{2}(2)(0.25^2)=0.0875$ | Gamma |
| 4 | $0.06-\frac{1}{2}(4)(0.08^2)=0.0472$ | $0.10-\frac{1}{2}(4)(0.15^2)=0.0550$ | $0.15-\frac{1}{2}(4)(0.25^2)=0.0250$ | Beta |
| 6 | $0.06-\frac{1}{2}(6)(0.08^2)=0.0408$ | $0.10-\frac{1}{2}(6)(0.15^2)=0.0325$ | $0.15-\frac{1}{2}(6)(0.25^2)=-0.0375$ | Alpha |

As risk aversion increases, the preferred portfolio shifts toward lower risk.

### Risk and Risk Aversion
*   Mean-Variance (M-V) Criterion
    *   Portfolio Alpha dominates Portfolio Beta if:
        $E(r_{\text{Alpha}}) \ge E(r_{\text{Beta}})$ and $\sigma_{\text{Alpha}} \le \sigma_{\text{Beta}}$ and at least one inequality is strict
*   Indifference Curves
    *   Equally preferred portfolios will lie in the mean-standard deviation plane on an indifference curve
    *   All portfolio points with the same utility value

1. Mean-Variance (M-V) Criterion — "Quy tắc so sánh áp đảo"
Đây là một quy tắc rất trực quan, không cần tính utility U gì cả:

Danh mục Alpha "áp đảo" (dominates) danh mục Beta nếu:

	* E(r_Alpha) ≥ E(r_Beta) — Alpha có lợi nhuận kỳ vọng bằng hoặc cao hơn
	* σ_Alpha ≤ σ_Beta — Alpha có rủi ro bằng hoặc thấp hơn
	* Và ít nhất một trong hai điều kiện trên phải là bất đẳng thức chặt (hơn hẳn, không chỉ bằng)


Nói đơn giản: nếu Alpha vừa lời nhiều hơn vừa an toàn hơn (hoặc bằng) Beta, thì không có lý do gì để chọn Beta cả — ai cũng sẽ chọn Alpha, bất kể mức độ ngại rủi ro A của họ là bao nhiêu.

→ Đây là trường hợp hiếm khi so sánh 2 danh mục bất kỳ (vì thường danh mục rủi ro cao hơn thì lời cũng cao hơn), nhưng khi nó xảy ra thì kết luận rất rõ ràng, không cần tranh cãi.

2. Indifference Curves — "Đường bàng quan" (đường thờ ơ)
Vấn đề đặt ra: Khi 2 danh mục không áp đảo lẫn nhau (ví dụ: Gamma lời nhiều hơn nhưng cũng rủi ro hơn Alpha), làm sao so sánh? → Đây là lúc cần đến utility U bạn học ở slide trước, và Indifference Curve là cách vẽ hình cho utility đó.

Ý nghĩa của đường cong trong hình:

	* Trục ngang (σ_p): độ rủi ro của danh mục
	* Trục dọc (E(r)): lợi nhuận kỳ vọng của danh mục
	* Đường cong xanh = tập hợp tất cả các điểm (σ, E(r)) cho cùng một giá trị U — tức cùng một mức độ "hấp dẫn" đối với nhà đầu tư đó.

Tại sao gọi là "bàng quan" (indifference)?
Vì nhà đầu tư thờ ơ, không thiên vị giữa điểm O và điểm P trên cùng đường cong — dù chúng có rủi ro và lợi nhuận khác nhau, nhưng utility U là bằng nhau, nên với nhà đầu tư đó, hai danh mục này "ngang giá trị" như nhau.

Vì sao đường cong dốc lên?
Vì nhà đầu tư ngại rủi ro (A > 0): nếu rủi ro σ tăng, thì để giữ U không đổi, lợi nhuận kỳ vọng E(r) phải tăng theo để bù đắp — đúng với công thức U = E(r) − ½Aσ² bạn đã học.

Tóm lại mối liên hệ giữa 2 phần:
	* M-V Criterion: dùng khi một danh mục rõ ràng tốt hơn ở cả 2 mặt (không cần biết A)
	* Indifference Curve: dùng khi phải đánh đổi giữa lời và rủi ro — lúc này câu trả lời "danh mục nào tốt hơn" sẽ phụ thuộc vào A của từng nhà đầu tư, y hệt bài học ở ví dụ Alpha/Beta/Gamma trước đó.

### Portfolio Return
*   Portfolio return is the weighted average of the assets' returns, using beginning-of-period portfolio weights:
    $$
    r_p = \sum_{n=1}^N w_nr_n.
    $$

An investor holds the following portfolio at the beginning of the period:

| Asset | Price | Units | Initial value | Weight | Total return | Return contribution | Ending value |
| :--- | ---: | ---: | ---: | ---: | ---: | ---: | ---: |
| Alpha | 50.00 | 80 | 4,000.00 | 40% | 10% | 4.00% | 4,400.00 |
| Beta | 100.00 | 30 | 3,000.00 | 30% | 5% | 1.50% | 3,150.00 |
| Gamma | 75.00 | 40 | 3,000.00 | 30% | -4% | -1.20% | 2,880.00 |
| Total | | | 10,000.00 | 100% | | 4.30% | 10,430.00 |

$$
r_p=(0.40)(0.10)+(0.30)(0.05)+(0.30)(-0.04)=0.043=4.30\%.
$$

The change in total portfolio value gives the same return:

$$
r_p=\frac{10{,}430-10{,}000}{10{,}000}=4.30\%.
$$

Ý tưởng cốt lõi
Lợi nhuận của danh mục = trung bình có trọng số (weighted average) của lợi nhuận từng tài sản, với trọng số là tỷ trọng đầu tư ban đầu


Công thức:

r_p =∑_(n=1)^Nw_nr_n
Trong đó:

	* w_n = tỷ trọng vốn đầu tư vào tài sản n (tính tại đầu kỳ)
	* r_n = lợi nhuận (return) của tài sản n trong kỳ đó

Áp dụng vào ví dụ trong bảng
Nhà đầu tư có 3 tài sản: Alpha, Beta, Gamma. Cách đọc bảng theo từng cột:

| Cột | Ý nghĩa |
|---|---|
| **Price × Units** | Giá mỗi cổ phiếu/tài sản nhân số lượng nắm giữ → ra **Initial value** |
| **Weight** | Tỷ trọng = Initial value của tài sản ÷ Tổng initial value (10,000) |
| **Total return** | % lời/lỗ của riêng tài sản đó trong kỳ (10%, 5%, -4%) |
| **Return contribution** | = Weight × Total return → đóng góp của tài sản đó vào lợi nhuận chung |
| **Ending value** | Giá trị tài sản đó vào cuối kỳ = Initial value × (1 + Total return) |

Ví dụ Alpha: Weight 40% × Total return 10% = 4.00% (return contribution)

Cộng dồn lại để ra lợi nhuận danh mục
r_p =(0.40)(0.10)+(0.30)(0.05)+(0.30)(−0.04)= 0.043 = 4.30%
→ Đơn giản là cộng tất cả "Return contribution" của 3 tài sản lại: 4.00% + 1.50% + (-1.20%) = 4.30%

Kiểm tra chéo (rất hay!)
Slide cho thấy ta có thể tính r_p theo 2 cách hoàn toàn khác nhau nhưng ra cùng một kết quả:

Cách 1 — Dùng trọng số × lợi nhuận từng tài sản (công thức ở trên) → 4.30%

Cách 2 — Nhìn vào tổng giá trị danh mục thay đổi thế nào:
r_p = (10,430 − 10,000)/10,000 = 4.30%

→ Đây chính là cách kiểm tra logic: dù tính theo từng tài sản riêng lẻ hay theo tổng giá trị danh mục, kết quả phải khớp nhau — vì về bản chất chúng là một.

Điểm mấu chốt cần nhớ

Công thức r_p = Σ w_n r_n chỉ đúng khi w_n là trọng số đầu kỳ (không đổi trong suốt kỳ đó) — đây là điều kiện quan trọng để công thức tuyến tính đơn giản này hoạt động chính xác.

Ở slide trước (cell 30), ta chỉ có 1 kết quả thực tế: Alpha lời 10%, Beta lời 5%, Gamma lỗ 4% → ra r_p = 4.30%.

Nhưng thực tế, trước khi thị trường diễn ra, ta không biết chắc điều gì sẽ xảy ra. Có thể là:

Công thức: Expected Return = trung bình có trọng số theo xác suất
E(r_p)=∑_(n=1)^Nw_nE(r_n)

Về bản chất, công thức này giống hệt công thức portfolio return ở slide trước — chỉ khác là bây giờ ta dùng E(r_n) (lợi nhuận kỳ vọng) thay vì r_n (lợi nhuận thực tế).

Đọc bảng theo 2 chiều
Chiều ngang (theo từng market state): Tính portfolio return nếu trạng thái đó xảy ra, dùng đúng công thức r_p = Σw_n r_n của slide trước.

	* Ví dụ Boom: (0.40)(0.20) + (0.30)(0.12) + (0.30)(0.18) = 17.00%

Chiều dọc (hàng cuối "Expected return"): Với mỗi tài sản, lấy trung bình có trọng số xác suất của các market state:

	* Alpha: E(r_Alpha) = 0.20(0.20) + 0.30(0.14) + 0.35(0.10) + 0.15(-0.08) = 10.50%

  * Tương tự cho Beta = 6.25%, Gamma = 3.70%

Điểm hay: Normal state tái hiện đúng ví dụ slide trước
Nhìn vào hàng Normal (xác suất 35%): Alpha 10%, Beta 5%, Gamma -4% — chính xác là bộ số liệu ở slide Portfolio Return trước đó, và portfolio return = 4.30% — khớp với kết quả đã tính!

→ Ý nghĩa: slide trước chỉ là một lát cắt (một kịch bản cụ thể) của bức tranh lớn hơn ở slide này.

Kết quả cuối: Expected Portfolio Return = 7.19%
E(r_p)= 0.40(0.1050)+ 0.30(0.0625)+ 0.30(0.0370)= 7.19%
Điểm mấu chốt cần nhớ

  * r_p (slide trước) = lợi nhuận đã biết, đã xảy ra → dùng để đánh giá hiệu suất quá khứ

  * E(r_p) (slide này) = lợi nhuận kỳ vọng, trước khi biết tương lai → dùng để ra quyết định đầu tư (vì ta không thể đầu tư dựa vào thứ chưa xảy ra!)

  * Dù là tính theo tài sản trước rồi mới ghép trọng số, hay tính theo state trước rồi mới lấy kỳ vọng — kết quả cuối cùng vẫn như nhau (đây chính là nội dung của đoạn code Python ngay bên dưới slide này, minh họa 2 cách tính tương đương).

```python
import numpy as np
import pandas as pd
from IPython.display import display

states = ['Boom', 'Growth', 'Normal', 'Recession']
probabilities = pd.Series([0.20, 0.30, 0.35, 0.15], index=states, name='Probability')

asset_returns = pd.DataFrame({
    'Alpha': [0.20, 0.14, 0.10, -0.08],
    'Beta': [0.12, 0.08, 0.05, -0.02],
    'Gamma': [0.18, 0.10, -0.04, -0.10]
}, index=states)

weights = pd.Series({'Alpha': 0.40, 'Beta': 0.30, 'Gamma': 0.30})

# Method 1: weight the assets' expected returns.
expected_asset_returns = asset_returns.mul(probabilities, axis=0).sum()
expected_by_assets = expected_asset_returns @ weights

# Method 2: calculate the portfolio return in each state, then take its expectation.
portfolio_returns = asset_returns @ weights
expected_by_states = probabilities @ portfolio_returns

state_table = pd.concat(
    [probabilities, asset_returns, portfolio_returns.rename('Portfolio')], axis=1
)
display(state_table.style.format('{:.2%}'))

print('Expected asset returns:')
for asset, expected_return in expected_asset_returns.items():
    print(f'  {asset}: {expected_return:.2%}')

print(f'\nExpected portfolio return from asset averages: {expected_by_assets:.3%}')
print(f'Expected portfolio return from state returns:  {expected_by_states:.3%}')

assert np.isclose(expected_by_assets, expected_by_states)
```

### Portfolio Expected Return: Two Equivalent Methods

*   Asset-first method: calculate each asset's expected return and then apply the portfolio weights
*   State-first method: calculate the portfolio return in each state and then apply the state probabilities
*   Both methods give the same result because portfolio return is linear:

$$
E(r_p)=\sum_n w_nE(r_n)=\sum_s p(s)r_p(s)=7.185\%\approx 7.19\%.
$$

### Portfolio Risk

Continue with Alpha and Beta from the state-based example in cell 31.

$$
\sigma^2(r_i) = p_1[r_1-E(r_i)]^2 + p_2[r_2-E(r_i)]^2 + \cdots + p_n[r_n-E(r_i)]^2
$$
$$
\sigma_i = \sqrt{\sigma^2(r_i)}
$$

| Market state | Prob. | Alpha ($r_{\mathrm{Alpha}}$) | $p[r_{\mathrm{Alpha}}-E(r_{\mathrm{Alpha}})]^2$ | Beta ($r_{\mathrm{Beta}}$) | $p[r_{\mathrm{Beta}}-E(r_{\mathrm{Beta}})]^2$ |
| :--- | ---: | ---: | ---: | ---: | ---: |
| Boom | 20% | 20.0% | 0.001805 | 12.0% | 0.000661 |
| Growth | 30% | 14.0% | 0.000368 | 8.0% | 0.000092 |
| Normal | 35% | 10.0% | 0.000009 | 5.0% | 0.000055 |
| Recession | 15% | -8.0% | 0.005134 | -2.0% | 0.001021 |
| | | $E(r)=10.50\%$ | $\sigma^2_{\mathrm{Alpha}}=0.007315$ | $E(r)=6.25\%$ | $\sigma^2_{\mathrm{Beta}}=0.001829$ |
| | | | $\sigma_{\mathrm{Alpha}}=8.55\%$ | | $\sigma_{\mathrm{Beta}}=4.28\%$ |

---


Slide này trả lời câu hỏi: "Rủi ro của một tài sản được đo lường chính xác như thế nào?" — và nó dùng chính bộ dữ liệu 4 market states bạn đã quen ở các slide trước để minh họa.

Ý tưởng cốt lõi: Rủi ro = Độ "lệch" so với kỳ vọng
Bạn đã biết E(r) là lợi nhuận kỳ vọng — tức giá trị "trung bình" mà ta dự đoán. Nhưng thực tế có thể lệch khỏi con số đó rất nhiều (lên Boom 20% hoặc xuống Recession -8%).

→ Variance (phương sai) σ² chính là cách đo "độ lệch trung bình bình phương" so với kỳ vọng — nói cách khác: thực tế thường xa kỳ vọng bao nhiêu?

Giải mã công thức
$$
\sigma^2(r_i) = p_1[r_1-E(r_i)]^2 + p_2[r_2-E(r_i)]^2 + \cdots + p_n[r_n-E(r_i)]^2
$$

Đọc từng phần:

	* [r_k − E(r_i)] → độ lệch (deviation) của lợi nhuận ở state k so với kỳ vọng
	* Bình phương lên → để loại dấu âm/dương (lệch lên hay lệch xuống đều tính là "rủi ro" như nhau), đồng thời phạt nặng hơn những độ lệch lớn
	* Nhân với p_k (xác suất) → state nào dễ xảy ra hơn thì đóng góp nhiều hơn vào rủi ro tổng
	* Cộng tất cả lại → ra variance, con số tổng hợp đo độ "phân tán" của lợi nhuận

Sau đó: σ_i = √(σ²(r_i)) → đưa về cùng đơn vị % với lợi nhuận (dễ so sánh và diễn giải hơn variance).

### Portfolio Risk

*   When calculating the expected return of the portfolio, we take the weighted average of the expected returns of the assets.

*   We cannot use the same method for variance or standard deviation (i.e., simply taking a weighted average of individual asset variances is incorrect due to the effect of covariances).

*   **Covariance:**
    *   Measures the directional relationship between the returns of two assets.
    *   At the portfolio level, the sign and magnitude of the covariances between pairs of assets significantly impact the overall volatility (risk) of the portfolio.

*   For two assets, $i$ and $j$, when considering $N$ possible states of the world (scenarios), each with a probability $p_k$:

    $ \text{Cov}(r_i, r_j) = p_1 \cdot [r_{i,1} - E(r_i)] \cdot [r_{j,1} - E(r_j)] + p_2 \cdot [r_{i,2} - E(r_i)] \cdot [r_{j,2} - E(r_j)] + \dots + p_N \cdot [r_{i,N} - E(r_i)] \cdot [r_{j,N} - E(r_j)] $

    $ \text{Cov}(r_i, r_j) = \sum_{k=1}^{N} p_k \cdot [r_{i,k} - E(r_i)] \cdot [r_{j,k} - E(r_j)] $
    
    Where:
    *   $\text{Cov}(r_i, r_j)$ is the covariance between the returns of asset $i$ and asset $j$.
    *   $N$ is the number of possible states/scenarios and $p_k$ is the probability of state/scenario $k$ occurring.
    *   $r_{i,k}$ is the return of asset $i$ if state/scenario $k$ occurs and $E(r_i)$ is the expected return of asset $i$.
    *   $r_{j,k}$ is the return of asset $j$ if state/scenario $k$ occurs and $E(r_j)$ is the expected return of asset $j$.

Covariance là gì? — Hiểu theo trực giác
Covariance đo lường: khi tài sản i tăng/giảm so với kỳ vọng của nó, thì tài sản j có xu hướng tăng/giảm so với kỳ vọng của nó theo cùng chiều hay ngược chiều?


	* Cov > 0: Hai tài sản có xu hướng di chuyển cùng chiều (cùng tăng, cùng giảm)
	* Cov < 0: Hai tài sản có xu hướng di chuyển ngược chiều (một tăng thì một giảm)
	* Cov ≈ 0: Hai tài sản không có mối liên hệ rõ ràng

Tại sao Covariance lại quan trọng ở cấp độ danh mục?
Đây là trực giác tài chính cốt lõi của đa dạng hóa (diversification):

	* Nếu 2 tài sản có Cov âm hoặc thấp → khi tài sản này giảm, tài sản kia có thể tăng (hoặc không giảm theo) → rủi ro tổng của danh mục được "triệt tiêu" bớt
	* Nếu 2 tài sản có Cov cao và dương → cả 2 cùng tăng cùng giảm → rủi ro không được giảm bớt chút nào, thậm chí còn khuếch đại

→ Đây chính là lý do câu "dấu và độ lớn của Cov quyết định mạnh mẽ đến biến động (risk) của cả danh mục" trong slide.

Giải mã công thức

$ \text{Cov}(r_i, r_j) = \sum_{k=1}^{N} p_k \cdot [r_{i,k} - E(r_i)] \cdot [r_{j,k} - E(r_j)] $

So sánh với công thức Variance ở slide trước:

$$
\sigma^2(r_i) = p_1[r_1-E(r_i)]^2 + p_2[r_2-E(r_i)]^2 + \cdots + p_n[r_n-E(r_i)]^2
$$

→ Bạn sẽ nhận ra: Variance chính là "Covariance của một tài sản với chính nó" — tức Cov(r_i, r_i) = σ²(r_i)! Vì khi i = j, công thức trở thành

[r_i − E(r_i)] × [r_i − E(r_i)] = [r_i − E(r_i)]².

Đọc từng phần của công thức Covariance:

	* [r_{i,k} − E(r_i)] → độ lệch của tài sản i tại state k so với kỳ vọng của nó
	* [r_{j,k} − E(r_j)] → độ lệch của tài sản j tại state k so với kỳ vọng của nó
	* Nhân 2 độ lệch này lại với nhau → nếu cả 2 cùng dương hoặc cùng âm (cùng chiều) → tích dương; nếu ngược chiều → tích âm
	* Nhân với p_k rồi cộng dồn qua các state → ra giá trị trung bình có xác suất của "độ đồng biến" này

```python
import numpy as np

# Alpha and Beta data from cell 31
probs = np.array([0.20, 0.30, 0.35, 0.15])
returns_alpha = np.array([0.20, 0.14, 0.10, -0.08])
returns_beta = np.array([0.12, 0.08, 0.05, -0.02])

E_alpha = probs @ returns_alpha
E_beta = probs @ returns_beta
var_alpha = probs @ (returns_alpha - E_alpha) ** 2
var_beta = probs @ (returns_beta - E_beta) ** 2
cov_alpha_beta = probs @ (
    (returns_alpha - E_alpha) * (returns_beta - E_beta)
)

# Two-asset portfolio calculations for different Alpha–Beta weights
weights = [(w_alpha, 1 - w_alpha) for w_alpha in np.linspace(1, 0, 11)]

print('w_Alpha  w_Beta  E(r_p)  Var(r_p)  Std(r_p)')
for w_alpha, w_beta in weights:
    E_rp = w_alpha * E_alpha + w_beta * E_beta
    var_p = (
        w_alpha**2 * var_alpha
        + w_beta**2 * var_beta
        + 2 * w_alpha * w_beta * cov_alpha_beta
    )
    std_p = np.sqrt(var_p)
    print(
        f'{w_alpha:7.0%} {w_beta:7.0%} '
        f'{E_rp:7.3%} {var_p:9.6f} {std_p:8.2%}'
    )
```

### Risk in Investments: Portfolio Risk

**Large portfolio variance**
$$ E(r_p) = \sum_{i=1}^{n} w_i E(r_i) \qquad \text{and} \qquad Var(r_p) = \sum_{i=1}^{n} \sum_{j=1}^{n} w_i w_j Cov(r_i, r_j) $$

**Diversification with an equally weighted portfolio**
$$ Var(r_p) = \frac{1}{n} (\text{Average Variance of the Individual Stocks}) + (1 - \frac{1}{n}) (\text{Average Covariance between the Stocks}) $$

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

# --- Input Parameter Widgets ---
style_desc = {'description_width': '200px'}
widget_layout_left = widgets.Layout(width='350px')
widget_layout_right = widgets.Layout(width='350px')

# --- Defaults from the Alpha–Beta example ---
er_a_widget = widgets.FloatText(value=0.1050, step=0.0001, description='Expected Return Alpha:', style=style_desc, layout=widget_layout_left)
std_a_widget = widgets.FloatText(value=0.085528, step=0.0001, description='Std. Dev. Alpha:', style=style_desc, layout=widget_layout_left)
corr_ab_widget = widgets.FloatText(value=0.9809, step=0.0001, min=-1, max=1, description='Correlation (Alpha, Beta):', style=style_desc, layout=widget_layout_left)
cov_ab_widget = widgets.FloatText(value=0.0035875, step=0.0000001, description='Covariance (Alpha, Beta):', style=style_desc, layout=widget_layout_left)

er_b_widget = widgets.FloatText(value=0.0625, step=0.0001, description='Expected Return Beta:', style=style_desc, layout=widget_layout_right)
std_b_widget = widgets.FloatText(value=0.042764, step=0.0001, description='Std. Dev. Beta:', style=style_desc, layout=widget_layout_right)
# --- End of example defaults ---

# --- Output Widgets ---
output_gmvp = widgets.Output()
output_table = widgets.Output()
output_plot = widgets.Output()

# --- Recalculate Button ---
recalculate_button = widgets.Button(
    description="Recalculate & Plot",
    button_style='info',
    tooltip='Click to update frontiers and table',
    icon='refresh',
    layout=widgets.Layout(width='auto', min_width='150px', margin='10px 0 0 0')
)

# --- Linking Logic for Correlation and Covariance ---
updating_internally = False

def format_float_widget_value(widget, value, precision):
    try:
        if pd.isna(value):
            actual_value_to_set = 0.0
        else:
            actual_value_to_set = float(f"{value:.{precision}f}")
        current_widget_val = widget.value
        if current_widget_val is None or abs(current_widget_val - actual_value_to_set) > 10**-(precision+1):
            widget.value = actual_value_to_set
    except Exception:
        pass

def update_covariance(change):
    global updating_internally
    if updating_internally: return
    updating_internally = True
    try:
        std_a = std_a_widget.value; std_b = std_b_widget.value; corr_ab = corr_ab_widget.value
        if std_a is not None and std_b is not None and corr_ab is not None and std_a >= 0 and std_b >= 0:
            new_cov = corr_ab * std_a * std_b
            format_float_widget_value(cov_ab_widget, new_cov, 6)
    except Exception: pass
    finally: updating_internally = False

def update_correlation(change):
    global updating_internally
    if updating_internally: return
    updating_internally = True
    try:
        std_a = std_a_widget.value; std_b = std_b_widget.value; cov_ab = cov_ab_widget.value
        if std_a is not None and std_b is not None and cov_ab is not None:
            if std_a > 1e-9 and std_b > 1e-9:
                corr_val = cov_ab / (std_a * std_b)
                format_float_widget_value(corr_ab_widget, np.clip(corr_val, -1, 1), 4)
            elif abs(std_a) < 1e-9 or abs(std_b) < 1e-9:
                 format_float_widget_value(corr_ab_widget, 0.0 if abs(cov_ab) < 1e-9 else np.nan, 4)
            else:
                 format_float_widget_value(corr_ab_widget, np.nan, 4)
    except Exception: pass
    finally: updating_internally = False

def on_std_dev_change(change):
    update_covariance(None)

corr_ab_widget.observe(update_covariance, names='value')
cov_ab_widget.observe(update_correlation, names='value')
std_a_widget.observe(on_std_dev_change, names='value')
std_b_widget.observe(on_std_dev_change, names='value')

# --- Core Calculation and Plotting Logic ---
def calculate_and_display_frontiers(b=None):
    with output_gmvp: clear_output(wait=True)
    with output_table: clear_output(wait=True)
    with output_plot: clear_output(wait=True)
    try:
        er_a = er_a_widget.value; std_a = std_a_widget.value; var_a = std_a**2
        er_b = er_b_widget.value; std_b = std_b_widget.value; var_b = std_b**2
        cov_ab = cov_ab_widget.value

        if any(v is None for v in [er_a, std_a, er_b, std_b, cov_ab]):
             with output_plot: print("Error: One or more input values are missing or invalid.")
             return

        plot_weights_a = np.linspace(0, 1, 100)
        plot_portfolio_returns = np.array([w_a * er_a + (1 - w_a) * er_b for w_a in plot_weights_a])
        plot_portfolio_variances = np.array([(w_a**2 * var_a) + ((1 - w_a)**2 * var_b) + (2 * w_a * (1 - w_a) * cov_ab) for w_a in plot_weights_a])
        plot_portfolio_variances = np.maximum(0, plot_portfolio_variances)
        plot_portfolio_std_devs = np.sqrt(plot_portfolio_variances)

        denominator_gmvp = var_a + var_b - 2 * cov_ab
        w_a_gmvp = 0.5 if abs(denominator_gmvp) < 1e-9 else (var_b - cov_ab) / denominator_gmvp
        w_a_gmvp = np.clip(w_a_gmvp, 0, 1); w_b_gmvp = 1 - w_a_gmvp
        gmvp_return = w_a_gmvp * er_a + w_b_gmvp * er_b
        gmvp_variance = (w_a_gmvp**2 * var_a) + (w_b_gmvp**2 * var_b) + (2 * w_a_gmvp * w_b_gmvp * cov_ab)
        gmvp_variance = max(0, gmvp_variance)
        gmvp_std_dev = np.sqrt(gmvp_variance)

        with output_gmvp:
            display(widgets.HTML("<h4>Long-Only Global Minimum Variance Portfolio (GMVP):</h4>"))
            gmvp_df = pd.DataFrame({
                'Weight Alpha': [f"{w_a_gmvp*100:.2f}%"], 'Weight Beta': [f"{w_b_gmvp*100:.2f}%"],
                'Expected Return E(Rp)': [f"{gmvp_return*100:.3f}%"], 'Variance (σp²)': [f"{gmvp_variance:.6f}"],
                'Std. Deviation (σp)': [f"{gmvp_std_dev*100:.2f}%"]})
            display(gmvp_df)

        table_weights_a = np.array([1.0, 0.9, 0.8, 0.7, 0.6, 0.5, 0.4, 0.3, 0.2, 0.1, 0.0])
        table_data = []
        for w_a_t in table_weights_a:
            w_b_t = 1 - w_a_t
            rp_t = w_a_t * er_a + w_b_t * er_b
            var_p_t = (w_a_t**2 * var_a) + (w_b_t**2 * var_b) + (2 * w_a_t * w_b_t * cov_ab)
            var_p_t = max(0, var_p_t)
            std_p_t = np.sqrt(var_p_t)
            table_data.append([f"{w_a_t*100:.0f}%", f"{w_b_t*100:.0f}%", f"{rp_t*100:.3f}%", f"{var_p_t:.6f}", f"{std_p_t*100:.2f}%"])
        df_table = pd.DataFrame(table_data, columns=["w_Alpha", "w_Beta", "E(rp)", "σp²", "σp"])
        with output_table:
            display(widgets.HTML("<h4>Portfolio Combinations Table:</h4>"))
            display(df_table)

        with output_plot:
            fig, ax = plt.subplots(figsize=(9, 6))
            ax.plot(plot_portfolio_std_devs, plot_portfolio_returns, linestyle='-', color='blue', label='Long-Only Opportunity Set')

            sort_indices = np.argsort(plot_portfolio_std_devs)
            sorted_stds = plot_portfolio_std_devs[sort_indices]; sorted_rets = plot_portfolio_returns[sort_indices]
            gmvp_sorted_idx = np.argmin(sorted_stds)

            eff_stds = sorted_stds[gmvp_sorted_idx:]; eff_rets = sorted_rets[gmvp_sorted_idx:]

            final_eff_stds = [eff_stds[0]] if len(eff_stds)>0 else []; final_eff_rets = [eff_rets[0]] if len(eff_rets)>0 else []
            if len(eff_stds) > 0:
                for k in range(1, len(eff_stds)):
                    if eff_rets[k] >= final_eff_rets[-1] - 1e-6:
                        final_eff_stds.append(eff_stds[k]); final_eff_rets.append(eff_rets[k])

            if final_eff_stds:
                 ax.plot(final_eff_stds, final_eff_rets, linestyle='-', color='green', linewidth=2.5, label='Efficient Frontier', zorder=3)

            ax.scatter([std_a], [er_a], color='red', marker='o', s=100, label='Alpha', zorder=4)
            ax.scatter([std_b], [er_b], color='purple', marker='o', s=100, label='Beta', zorder=4)
            ax.scatter([gmvp_std_dev], [gmvp_return], color='black', marker='*', s=200, label='GMVP', zorder=5)
            ax.set_title('Long-Only Portfolios: Minimum Variance & Efficient Frontier'); ax.set_xlabel('Portfolio Std Dev (σp)'); ax.set_ylabel('Portfolio Exp Return (E(rp))')
            ax.yaxis.set_major_formatter(plt.FuncFormatter(lambda y, _: f'{y*100:.2f}%')); ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, _: f'{x*100:.2f}%'))
            ax.legend(); ax.grid(True, linestyle='--', alpha=0.7); plt.tight_layout(); plt.show()
    except Exception as e:
        with output_gmvp: clear_output(wait=True); print(f"Error in GMVP: {e}")
        with output_table: clear_output(wait=True); print(f"Error in Table: {e}")
        with output_plot: clear_output(wait=True); print(f"Error in Plot: {e}")

# --- Linking Button and Initial Display ---
recalculate_button.on_click(calculate_and_display_frontiers)

title_widget = widgets.HTML("<h2>Portfolio Risk Calculation: Alpha and Beta (Long-Only)</h2>", layout=widgets.Layout(width='100%'))
formula_cov_widget = widgets.HTML("For a two-asset portfolio (using Covariance):<br>    σ²(rp) = w₁²σ²(r₁) + w₂²σ²(r₂) + 2w₁w₂Cov(r₁, r₂)", layout=widgets.Layout(margin='0 0 2px 0'))
formula_corr_widget = widgets.HTML("For a two-asset portfolio (using Correlation):<br>    σ²(rp) = w₁²σ²(r₁) + w₂²σ²(r₂) + 2w₁w₂ρ₁₂σ₁σ₂", layout=widgets.Layout(margin='0 0 15px 0'))
inputs_col_a_corr_cov = widgets.VBox([er_a_widget, std_a_widget, corr_ab_widget, cov_ab_widget])
inputs_col_b = widgets.VBox([er_b_widget, std_b_widget], layout=widgets.Layout(margin='0 0 0 20px'))
left_panel_with_button = widgets.VBox([inputs_col_a_corr_cov, recalculate_button])
input_ui_grid = widgets.HBox([left_panel_with_button, inputs_col_b])

full_ui = widgets.VBox([
    title_widget, formula_cov_widget, formula_corr_widget, input_ui_grid,
    output_gmvp, widgets.HTML("<hr>"), output_table, widgets.HTML("<hr>"), output_plot
])

display(full_ui)
calculate_and_display_frontiers()
```

### Opportunity Set of Risky Assets
*   The minimum-variance frontier shows the risk–return opportunities available to the investor.

*   The efficient frontier contains the portfolio with the highest expected return at each risk level.

*   Capital allocation combines risky portfolio Alpha with a risk-free asset.
    *   If Gamma invests fraction $y$ in Alpha and $1-y$ in the risk-free asset:
        $$E(r_{\text{Gamma}})=yE(r_{\text{Alpha}})+(1-y)r_f$$
        $$\sigma_{\text{Gamma}}=y\sigma_{\text{Alpha}}, \qquad 0\leq y\leq1$$
    *   Changing $y$ determines Gamma's risk and expected return.
