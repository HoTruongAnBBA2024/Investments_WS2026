**1. Môi trường đầu tư (The Investment Environment: Asset Classes & Financial Instruments)**

* Liên hệ thực tế: Bạn muốn mở một quán cà phê (tài sản thực), nhưng thay vì tự làm, bạn quyết định mua cổ phiếu của Highlands Coffee hoặc gửi tiết kiệm vào ngân hàng (tài sản tài chính).

**2. Chứng khoán thu nhập cố định (Fixed-Income Securities)**
* Nội dung sơ lược: Tập trung vào các công cụ như Trái phiếu (bonds) - nơi bạn đóng vai trò là người cho vay và nhận lại một khoản lãi suất cố định, ít rủi ro hơn cổ phiếu.
* Liên hệ thực tế: Khi chính phủ hoặc các tập đoàn lớn cần tiền làm đường cao tốc hay xây nhà máy, họ phát hành trái phiếu. Bạn mua trái phiếu của họ để nhận lãi định kỳ hàng năm ổn định hơn gửi tiết kiệm.
* Mối liên kết: Sau khi biết các loại tài sản ở phần 1, bạn đi sâu vào phân tích nhóm tài sản có độ an toàn cao nhất trước (Trái phiếu), làm nền tảng định giá cho các tài sản rủi ro hơn phía sau.

**3. Lý thuyết và thực hành danh mục đầu tư (Portfolio Theory and Practice)**
* Nội dung sơ lược: Người xưa có câu "Không bỏ tất cả trứng vào một giỏ". Phần này dạy bạn cách kết hợp các tài sản (cổ phiếu, trái phiếu) lại với nhau để giảm thiểu rủi ro tối đa nhưng vẫn đạt được lợi nhuận mong muốn.
* Liên hệ thực tế: Thay vì dồn hết tiền mua một cổ phiếu công nghệ đầy biến động, bạn chia tiền ra: 60% vào cổ phiếu công nghệ, 30% vào cổ phiếu ngành điện nước ổn định, và 10% vào trái phiếu chính phủ.
* Mối liên kết: Từ việc hiểu rõ từng công cụ riêng lẻ ở phần 1 & 2, bạn bắt đầu học cách "trộn" chúng lại với nhau thành một danh mục tối ưu.

**4. Các mô hình định giá tài sản (Asset Pricing Models: CAPM, APT, Multifactor)**
* Nội dung sơ lược: Làm sao để biết một cổ phiếu có mức giá hợp lý hay chưa dựa trên mức độ rủi ro của nó? Các mô hình này (như CAPM) giúp bạn tính toán mối quan hệ giữa rủi ro hệ thống của thị trường và lợi nhuận kỳ vọng.
* Liên hệ thực tế: Nếu bạn muốn đầu tư vào một cổ phiếu công nghệ có độ biến động gấp đôi thị trường, mô hình này sẽ cho bạn biết bạn xứng đáng nhận được mức lợi nhuận tối thiểu là bao nhiêu để bù đắp cho sự "đau tim" đó.
* Mối liên kết: Sau khi đã biết cách xây dựng danh mục ở phần 3, bạn cần một công cụ toán học và lý thuyết chuẩn chỉnh để định giá chính xác từng tài sản cấu thành nên danh mục đó.

**5. Giả thuyết thị trường hiệu quả & Tài chính hành vi (Efficient Market Hypothesis & Behavioral Finance)**
* Nội dung sơ lược: Thị trường có luôn phản ánh đúng giá trị thực hay không? Bạn sẽ học về cuộc đối đầu giữa lý thuyết "thị trường luôn đúng và phản ánh mọi thông tin tức thời" và thực tế "con người thường quyết định dựa trên cảm xúc, tâm lý đám đông".
* Liên hệ thực tế: Các đợt sốt đất hay bong bóng chứng khoán, khi mọi người đổ xô đi mua chỉ vì sợ bỏ lỡ cơ hội (FOMO), chính là minh họa rõ nét của Tài chính hành vi.
* Mối liên kết: Khi áp dụng các mô hình định giá ở phần 4, bạn nhận ra thị trường không phải lúc nào cũng hoạt động hoàn hảo như công thức. Phần này giúp bạn hiểu được các yếu tố tâm lý con người tác động thế nào đến giá cả thực tế.

**6. Công cụ phái sinh (Derivatives: Options, Futures, Forwards, Swaps)**
* Nội dung sơ lược: Các công cụ mà giá trị của chúng phụ thuộc vào một tài sản cơ sở khác. Chúng được sinh ra ban đầu để phòng ngừa rủi ro (bảo hiểm), nhưng cũng là công cụ đầu cơ mạnh mẽ.

* Mối liên kết: Đây là đỉnh cao của mạng lưới. Khi bạn đã nắm rõ về tài sản cơ sở (cổ phiếu ở phần 1, trái phiếu ở phần 2), hiểu về rủi ro ở phần 3, 4 và tâm lý thị trường ở phần 5, bạn sẽ sử dụng Phái sinh như một công cụ tối thượng để bảo vệ danh mục của mình hoặc tìm kiếm lợi nhuận trong mọi điều kiện thị trường (ngay cả khi thị trường đi xuống).

**1. Tài sản thực (Real Assets) vs. Tài sản tài chính (Financial Assets)**

* **Tài sản thực (Real Assets)**: Là những thứ có "hình hài" vật lý hoặc sở hữu trí tuệ, trực tiếp tham gia vào quá trình sản xuất ra của cải, dịch vụ cho xã hội.
Ví dụ sinh viên: Chiếc laptop bạn dùng để học tập/làm thêm, chiếc xe máy bạn đi học, hay căn nhà trọ, máy pha cà phê của một quán nước. Nếu không có chúng, chúng ta không tạo ra sản phẩm hay dịch vụ trực tiếp được.
* **Tài sản tài chính (Financial Assets)**: Là những "tờ giấy" (hoặc dòng dữ liệu trên app) đại diện cho quyền sở hữu hoặc quyền đòi nợ đối với tài sản thực. Bản thân chúng không trực tiếp làm ra của cải, nhưng chúng giúp phân bổ nguồn vốn.
* **Ví dụ sinh viên**: Khi bạn gửi tiết kiệm vào ngân hàng để lấy lãi, chiếc thẻ tiết kiệm đó là tài sản tài chính. Hoặc khi bạn dùng app điện thoại mua một vài cổ phiếu của FPT hay Vinamilk, bạn đang nắm giữ tài sản tài chính (quyền được chia lợi nhuận từ những máy móc, nhà xưởng thực tế của họ).
* **Câu hỏi nhỏ để bạn suy ngẫm**: Nếu bạn tự mở một tiệm trà sữa, thì máy dập nắp cốc và số tiền bạn góp chung với bạn thân (có giấy cam kết chia đôi lợi nhuận), đâu là tài sản thực và đâu là tài sản tài chính?

**2. Vai trò của thị trường tài chính đối với nền kinh tế**

Thị trường tài chính không chỉ là nơi mua bán kiếm lời, mà nó đóng vai trò cực kỳ quan trọng:

* **Truyền tải thông tin (The Informational Role)**: Giá cổ phiếu tăng/giảm phản ánh đánh giá chung của xã hội về tương lai của doanh nghiệp.

Ví dụ: Nếu mọi người tin rằng xe điện là tương lai, dòng tiền đổ vào mua cổ phiếu Tesla sẽ làm giá nó tăng vọt, báo hiệu cho các nhà quản lý biết nên tập trung nguồn lực vào đâu.

* **Chuyển dịch thời gian tiêu dùng (Consumption Timing)**: Bạn kiếm được tiền hôm nay nhưng chưa muốn tiêu hết, bạn muốn dành cho lúc nghỉ hưu hoặc mua nhà sau này. Tài sản tài chính giúp bạn gửi gắm dòng tiền đó để nó sinh sôi.

Ví dụ: Tiền lì xì hay tiền lương làm thêm của bạn nếu cất dưới gối sẽ bị mất giá (lạm phát). Bạn đem mua chứng chỉ quỹ hoặc gửi tiết kiệm chính là đang chuyển tiêu dùng từ hiện tại sang tương lai.

* **Phân bổ rủi ro (Allocation of Risk)**: Mỗi người có mức độ "chịu nhiệt" rủi ro khác nhau. Thị trường chia nhỏ rủi ro ra thành nhiều mức độ để ai cũng có thể tham gia.

Ví dụ: Bạn thích an toàn? Bạn mua Trái phiếu (chắc chắn nhận lãi cố định). Bạn thích mạo hiểm để mong lời nhiều? Bạn mua Cổ phiếu.

* **Quản trị doanh nghiệp (Corporate Governance)**: Các cổ đông lớn và nhà phân tích luôn theo dõi sát sao ban giám đốc để đảm bảo họ không làm ăn gian dối hay lãng phí tiền của nhà đầu tư.

   * Giải thích: Thị trường đóng vai trò như một bộ lọc thông minh. Những công ty có ý tưởng kinh doanh tiềm năng và quản lý tốt sẽ thu hút được nhiều nhà đầu tư rót vốn, ngược lại các dự án kém khả thi sẽ bị đào thải.
   * Thị trường tạo điều kiện thuận lợi cho việc chuyển giao vốn từ những người muốn tìm kiếm lợi nhuận từ tài sản của họ sang những người cần vốn để mở rộng quy mô.
   * Giải thích: Đây là cầu nối trung gian. Một bên là những người có tiền nhàn rỗi muốn đầu tư sinh lời (như cá nhân gửi tiết kiệm, mua chứng khoán), và một bên là các doanh nghiệp đang cần tiền để sản xuất kinh doanh.
   * Nguồn vốn, được huy động thông qua vốn cổ phần (vốn chủ sở hữu) và vốn nợ, có thể được sử dụng để:
   * phát triển doanh nghiệp,
      * tài trợ cho các khoản đầu tư vào bất động sản, thiết bị và công nghệ mới,
      * cấp vốn cho các dự án cơ sở hạ tầng.
      * Giải thích: Doanh nghiệp có thể gọi vốn bằng cách bán cổ phiếu (vốn cổ phần) hoặc vay tiền/phát hành trái phiếu (vốn nợ). Số tiền này sẽ đem đi tái đầu tư để mở rộng nhà xưởng, mua máy móc hiện đại hoặc xây cầu đường, bến cảng.
   * Các cá nhân và doanh nghiệp có thể đầu tư vào các loại chứng khoán để tạo ra tài sản (gia tăng sự giàu có).
   * Giải thích: Bằng cách mua các loại chứng khoán (như cổ phiếu, trái phiếu), người dân và các tổ chức có cơ hội chia sẻ lợi nhuận cùng doanh nghiệp, từ đó giúp số vốn ban đầu của mình sinh sôi nảy nở theo thời gian.

**3. Quy trình đầu tư (The Investment Process)**
Khi bắt đầu đầu tư, bạn sẽ trải qua hai bước lớn:

* **Phân bổ tài sản (Asset Allocation)**: Bạn quyết định chia tiền của mình vào các giỏ lớn nào (bao nhiêu % vào cổ phiếu, bao nhiêu % gửi tiết kiệm, bao nhiêu % mua vàng).
* **Lựa chọn chứng khoán (Security Selection)**: Trong giỏ cổ phiếu đã chọn ở bước 1, bạn sẽ chọn cụ thể mã cổ phiếu nào (FPT, MWG hay VNM?).

**Cách tiếp cận từ trên xuống (Top-down)**: Bạn nhìn bức tranh kinh tế vĩ mô trước (ngành nào đang hưởng lợi? -> chọn ngành đó -> rồi mới chọn công ty tốt nhất trong ngành).

**Cách tiếp cận từ dưới lên (Bottom-up)**: Bạn không quan tâm lắm đến vĩ mô, bạn chỉ đi tìm những doanh nghiệp cực tốt, có giá cổ phiếu đang quá rẻ để mua trực tiếp.

### The Investment Environment
* Competitive Markets
  * Risk-Return Trade-Off
    * Higher expected returns compensate for risk that cannot be diversified away.
    * Higher risk does not guarantee higher returns.
  * Efficient Markets
    * Prices quickly adjust to all relevant information
    * There should be neither underpriced nor overpriced securities
  * Passive Management
    * Holding a highly diversified portfolio
    * No attempt to find undervalued securities
    * No attempt to time the market
  * Active Management
    * Finding mispriced securities
    * Timing the market

---



4. Thị trường cạnh tranh và ranh giới giữa Chủ động & Thụ động

Đánh đổi Rủi ro - Lợi nhuận (Risk-Return Trade-Off): Không có chuyện "việc nhẹ lương cao". Muốn có cơ hội nhận lợi nhuận cao hơn, bạn buộc phải chấp nhận rủi ro lớn hơn. Tuy nhiên, rủi ro cao chưa chắc đã mang lại lợi nhuận cao nếu bạn đầu tư sai cách.

Thị trường hiệu quả (Efficient Markets): Giá cả trên thị trường phản ánh rất nhanh mọi thông tin công bố. Vì vậy, việc tìm được một món hàng "quá hời" hoặc "quá đắt" là rất khó.

Vì giá cả luôn bám rất sát và thay đổi theo thông tin mới nhất, nên mức giá bạn nhìn thấy trên bảng điện tử tại bất kỳ thời điểm nào đều được coi là "mức giá hợp lý nhất" dựa trên tất cả những gì thế giới đã biết về tài sản đó. Do đó, bạn rất khó mua được tài sản với giá rẻ hơn nhiều so với giá trị thực của nó, và ngược lại.

Quản lý thụ động (Passive Management): Chấp nhận thị trường luôn đúng, chỉ cần mua một danh mục đa dạng đại diện cho toàn thị trường (như quỹ chỉ số ETF) và nắm giữ lâu dài.

Quản lý chủ động (Active Management): Cố gắng tìm ra những cổ phiếu bị định giá sai để mua vào/bán ra nhằm "đánh bại" thị trường.

### Asset Classes and Financial Instruments: Derivatives
* Options
  * A call option gives its holder the right to purchase an asset for a specified price, called the exercise or strike price, on or before a specified expiration date.

* Futures/Forwards
  * A futures contract calls for delivery of an asset (or, in some cases, its cash value) at a specified delivery or maturity date for an agreed-upon price, called the futures price, to be paid at contract maturity.

    * The party who takes the short position commits to delivering the asset at contract maturity.

## I. FIXED INCOME (TÀI SẢN THU NHẬP CỐ ĐỊNH)
## 1. Bản chất bản chất và đặc tính

* Mối quan hệ: Đây là quan hệ Chủ nợ và Con nợ. Khi bạn mua một công cụ thu nhập cố định, bạn đang cho tổ chức phát hành (Chính phủ hoặc Doanh nghiệp) vay tiền.
* Dòng tiền: Được xác định trước một cách rõ ràng. Tổ chức phát hành có nghĩa vụ pháp lý phải trả một mức lãi suất định kỳ (gọi là Coupon) và hoàn trả lại số vốn gốc (Mệnh giá) khi đến hạn (Đáo hạn).
* Quyền lợi: Người nắm giữ không có quyền biểu quyết, không được tham gia vào quyết định kinh doanh của doanh nghiệp. Tuy nhiên, nếu doanh nghiệp phá sản, chủ nợ là đối tượng được ưu tiên trả nợ trước cổ đông.
* Rủi ro: Thấp hơn cổ phiếu. Rủi ro lớn nhất là rủi ro vỡ nợ (tổ chức phát hành không có khả năng trả lãi/gốc) và rủi ro lạm phát (khi lạm phát cao hơn mức lãi suất cố định bạn nhận được).

## 2. Các công cụ phổ biến

* Trái phiếu Chính phủ (Sovereign Bonds): Do Bộ Tài chính phát hành để bù đắp ngân sách. Được coi là phi rủi ro vỡ nợ.
* Trái phiếu Doanh nghiệp (Corporate Bonds): Do các công ty phát hành để tài trợ dự án. Lãi suất cao hơn trái phiếu chính phủ do có rủi ro kinh doanh.
* Chứng chỉ tiền gửi (CD) hoặc Tiền gửi tiết kiệm ngân hàng.

## 3. Ví dụ vận dụng cho sinh viên BBA
Giả sử trường đại học của bạn đang cần 500 tỷ đồng để xây dựng một khu phức hợp Ký túc xá - Thư viện thông minh mới. Trường không muốn chia sẻ quyền sở hữu trường cho ai, nên họ phát hành Trái phiếu Nhà trường kỳ hạn 5 năm với lãi suất 8%/năm.
Bạn là sinh viên BBA có quỹ tiết kiệm từ việc đi làm thêm, bạn mua 10 triệu đồng trái phiếu này.

* Vận dụng: Bạn đóng vai trò là chủ nợ của trường. Mỗi năm trường bắt buộc phải trả cho bạn 800.000 đồng tiền lãi, bất kể năm đó trường tuyển sinh được nhiều hay ít sinh viên. Sau 5 năm, trường trả lại cho bạn đúng 10 triệu đồng gốc.

## II. EQUITY (VỐN CỔ PHẦN / CỔ PHIẾU)
## 1. Bản chất bản chất và đặc tính

* Mối quan hệ: Đây là quan hệ Đồng sở hữu. Khi bạn mua cổ phiếu của một công ty, bạn trở thành chủ sở hữu một phần doanh nghiệp đó (Cổ đông).
* Dòng tiền: Không cố định và không có gì đảm bảo. Bạn kiếm tiền từ hai nguồn: Cổ tức (phần lợi nhuận sau thuế được chia, tùy thuộc vào kết quả kinh doanh và quyết định của Đại hội cổ đông) và Thặng dư vốn (giá cổ phiếu tăng lên trên thị trường).
* Quyền lợi: Bạn có quyền biểu quyết tại Đại hội đồng cổ đông, có quyền bầu Hội đồng quản trị và can thiệp vào chiến lược doanh nghiệp (tương ứng với tỷ lệ cổ phần nắm giữ).
* Rủi ro: Cao. Nếu công ty làm ăn thua lỗ, giá cổ phiếu có thể về bằng 0 và bạn mất trắng. Khi phá sản, cổ đông là những người cuối cùng nhận được phần tài sản còn lại (nếu có).

## 2. Các công cụ phổ biến

* Cổ phiếu phổ thông (Common Stock): Loại phổ biến nhất, có quyền biểu quyết và nhận cổ tức biến động.
* Cổ phiếu ưu đãi (Preferred Stock): Loại lai ghép, không có quyền biểu quyết nhưng được ưu tiên nhận cổ tức cố định trước cổ phiếu phổ thông.

## 3. Ví dụ vận dụng cho sinh viên BBA
Một nhóm sinh viên khóa trên của khoa BBA thành lập một startup phát triển ứng dụng giao đồ ăn nội khu ký túc xá. Họ cần vốn để lập trình app và marketing nên họ chia công ty thành 100.000 cổ phần. Bạn dùng số tiền tích lũy mua 1.000 cổ phần (sở hữu 1% công ty).

* Vận dụng: Bạn không phải chủ nợ. Nếu ứng dụng bùng nổ, doanh thu tăng trưởng vượt bậc, giá trị 1% của bạn có thể tăng từ 10 triệu lên 100 triệu đồng, và bạn được chia cổ tức rất lớn. Ngược lại, nếu dự án thất bại sau 3 tháng, bạn chấp nhận mất trắng số tiền đầu tư mà không có quyền bắt nhóm founder phải trả lại tiền cho bạn.

## III. DERIVATIVES (CÔNG CỤ PHÁI SINH)


  ## 1. Bản chất bản chất và đặc tính

  * Bản chất: Đây không phải là tài sản sở hữu trực tiếp, mà là một Hợp đồng tài chính giữa hai hoặc nhiều bên. Giá trị của hợp đồng này được phái sinh (bắt nguồn) từ giá trị của một tài sản cơ sở bên dưới (tài sản cơ sở có thể là cổ phiếu, trái phiếu, hàng hóa như vàng, dầu, hoặc chỉ số thị trường).
  
  * Mục đích sử dụng: Có hai mục đích chính trong quản trị và đầu tư:
  
  * Phòng vệ rủi ro (Hedging): Bảo vệ bản thân trước sự biến động giá bất lợi trong tương lai.
   * Đầu cơ (Speculation): Đặt cược vào xu hướng giá để kiếm lợi nhuận nhanh (thường dùng đòn bẩy rất cao).
   * Đặc tính dòng tiền: Lãi/lỗ phụ thuộc vào sự chênh lệch giữa giá thỏa thuận trên hợp đồng và giá thực tế của tài sản cơ sở tại thời điểm đáo hạn hoặc thực hiện quyền.

   ## 2. Các công cụ phổ biến

   * Hợp đồng kỳ hạn (Forwards) / Hợp đồng tương lai (Futures): Bắt buộc hai bên phải mua/bán tài sản tại một thời điểm tương lai với giá cố định từ trước.
   * Hợp đồng quyền chọn (Options): Người mua trả một khoản phí để có quyền (nhưng không bắt buộc) mua hoặc bán tài sản với giá cố định.

   ## 3. Ví dụ vận dụng cho sinh viên BBA
   Câu lạc bộ (CLB) Sinh viên Nghiên cứu Khoa học của bạn lên kế hoạch tổ chức một đêm Gala lớn vào cuối học kỳ. Họ cần mua 50 thùng nước ngọt của một đại lý vào ngày diễn ra sự kiện (3 tháng nữa). Hiện tại giá là 200.000đ/thùng. Tuy nhiên, CLB lo ngại 3 tháng nữa vào mùa lễ hội, giá nước ngọt sẽ tăng lên 300.000đ/thùng, làm vỡ ngân sách của CLB.

   * Vận dụng phòng vệ rủi ro (Hedging): Chủ tịch CLB (một sinh viên BBA) đến gặp chủ đại lý nước ngọt và ký một Hợp đồng kỳ hạn (Forward Contract): Thỏa thuận đúng 3 tháng nữa CLB sẽ mua 50 thùng nước với giá cố định là 210.000đ/thùng.
   * Kịch bản 1: Đến ngày diễn ra, giá nước tăng vọt lên 300.000đ/thùng. Nhờ hợp đồng phái sinh này, CLB vẫn được mua với giá 210.000đ. CLB đã phòng vệ rủi ro thành công (tiết kiệm được khoản chi phí lớn).
      * Kịch bản 2: Đến ngày đó, đại lý giảm giá còn 150.000đ/thùng. Vì đây là hợp đồng kỳ hạn có tính bắt buộc, CLB vẫn phải ngậm ngùi mua với giá 210.000đ.

### Asset Classes and Financial Instruments

Zeta is a biotechnology startup. It owns laboratory equipment worth \$100,000 and holds \$60,000 in cash contributed by its founders.

For each transaction below, identify the real and financial assets involved. Explain whether financial claims are created, transferred, or extinguished. Ignore interest, taxes, and transaction costs.

a. Zeta issues five-year corporate bonds with a total face value of \$250,000 and receives \$250,000 in cash.

b. Zeta uses the bond proceeds and \$40,000 of its existing cash to develop a new drug, obtaining a patent for the formula.

c. Zeta transfers the patent to Eta, a pharmaceutical company. In exchange, Eta issues 6,000 new common shares to Zeta.

d. Zeta sells all 6,000 Eta shares to other investors for \$50 per share. It uses part of the proceeds to redeem all its outstanding bonds at face value.
