## Lecture 11: Binomial and Black-Scholes Valuation

Unless stated otherwise, options are European, cover one share and have non-dividend-paying stocks as their underlying. Ignore taxes and transaction costs. Fractional positions, short sales and risk-free borrowing and lending are allowed unless a question restricts them. Short-sale proceeds are available for investment.

For Black-Scholes calculations, rates are continuously compounded and volatility is annualized. The standard normal CDF may be evaluated with a calculator or Excel's `NORM.S.DIST(d,TRUE)`. Retain unrounded values in intermediate calculations.
